# Statistical Tests

## Introduction

Statistical tests are mathematical methods used to determine whether there is enough evidence in sample data to make a conclusion about a population.

Different statistical tests are used for different types of data and different situations.

The main statistical tests covered in this notebook are:

- T-Test
- Z-Test
- Chi-Square Test
- ANOVA
- Mann-Whitney U Test
- Shapiro-Wilk Test

The choice of test depends on:

- Type of data
- Number of groups
- Sample size
- Data distribution
- Whether the data is independent or paired
- Statistical assumptions

# 1. T-Test
### Definition

A T-Test is a statistical test used to determine whether there is a significant difference between means.

It is commonly used when the sample size is relatively small and the population standard deviation is unknown.

There are different types of T-Tests:

### One-sample T-Test
- Independent two-sample T-Test
- Paired T-Test
- One-Sample T-Test

Compares a sample mean with a known or hypothesized population mean.

### Independent T-Test

Compares the means of two independent groups.

### Paired T-Test

Compares two related measurements from the same subjects.

###Formula

For a one-sample T-Test:

t = (x̄ - μ) / (s / √n)

Where:

- x̄ = sample mean
- μ = hypothesized population mean
- s = sample standard deviation
- n = sample size

### Assumptions
- Data should be numerical.
- Observations should be independent.
- Data should be approximately normally distributed for small samples.
- For an independent T-Test, the two groups should be independent.
- For the standard independent T-Test, variances should be reasonably similar.

### When to Use

Use a T-Test when:

- Comparing means.
- Population standard deviation is unknown.
- Sample size is relatively small.
- Data is approximately normally distributed.
### When Not to Use

 Do not use a T-Test when:

- Data is strongly non-normal and sample size is small.
- Data is categorical.
- More than two independent groups need to be compared.
- Observations are not independent.

For non-normal data, a non-parametric test such as the Mann-Whitney U Test may be more appropriate.

### Simple Example

Suppose the average delivery time claimed by a company is 30 minutes.

We collect a sample of delivery times and want to check whether the actual average is significantly different from 30 minutes.

H₀: μ = 30

H₁: μ ≠ 30

### Python Example

```python
import numpy as np
from scipy.stats import ttest_1samp

delivery_times = np.array([28, 30, 31, 32, 29, 33, 30, 31])

t_statistic, p_value = ttest_1samp(delivery_times, 30)

print("T-Statistic:", t_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example

A company claims that the average customer waiting time is 30 minutes.

The company collects sample data and uses a T-Test to determine whether the actual average waiting time is significantly different from 30 minutes.

### AI/ML Example

A T-Test can be used to compare the average prediction errors of two models when the assumptions for the test are appropriate.

### Interpretation

If:

p-value ≤ 0.05

we reject the null hypothesis and conclude that there is statistically significant evidence of a difference.

If:

p-value > 0.05

we fail to reject the null hypothesis.

## 2. Z-Test
### Definition

A Z-Test is a statistical test used to determine whether a sample mean is significantly different from a population mean or whether two means are significantly different.

It is commonly used when the population standard deviation is known or when the sample size is sufficiently large.

### Formula

For a one-sample Z-Test:

Z = (x̄ - μ) / (σ / √n)

Where:

- x̄ = sample mean
- μ = population mean
- σ = population standard deviation
- n = sample size
- 
### Assumptions
- Data should be numerical.
- Observations should be independent.
- Population standard deviation should be known for the classical one-sample Z-Test.
- Data should be approximately normal, or the sample size should be large enough for the normal approximation to be reasonable.

### When to Use

Use a Z-Test when:

- Population standard deviation is known.
- Sample size is large.
- Comparing means or proportions under appropriate assumptions.
### When Not to Use

Do not use a Z-Test when:

- Population standard deviation is unknown and the sample is small.
- Data is categorical when testing means.
- Observations are dependent.
- The assumptions of the test are not satisfied.
### Simple Example

A factory says that the average weight of a product is 500 grams.

The population standard deviation is known to be 20 grams.

A sample is collected to test whether the average weight is significantly different from 500 grams.

### Python Example

```python
import numpy as np
from scipy.stats import norm

data = np.array([510, 505, 498, 512, 507, 503, 509, 511, 506, 508])

sample_mean = np.mean(data)
population_mean = 500
population_std = 20
n = len(data)

z_statistic = (sample_mean - population_mean) / (population_std / np.sqrt(n))

p_value = 2 * (1 - norm.cdf(abs(z_statistic)))

print("Sample Mean:", sample_mean)
print("Z-Statistic:", z_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example

A manufacturing company wants to check whether the average weight of its products is different from the specified target weight.

### AI/ML Example

Z-tests can be used for statistical comparisons when large samples and known population variance assumptions are appropriate.

### Interpretation

A large absolute Z-score indicates that the observed sample mean is far from the hypothesized population mean.

A small p-value indicates evidence against the null hypothesis.



## 3. Chi-Square Test
### Definition

The Chi-Square Test is a statistical test used mainly for categorical data.

It compares observed frequencies with expected frequencies.

There are two common types:

- Chi-Square Goodness-of-Fit Test
- Chi-Square Test of Independence
### Goodness-of-Fit

Checks whether observed category frequencies match expected frequencies.

### Test of Independence

Checks whether two categorical variables are associated with each other.

Formula

χ² = Σ [(O - E)² / E]

Where:

- O = Observed frequency
- E = Expected frequency
\chi^2 = \sum \frac{(O_i-E_i)^2}{E_i}
### Assumptions
- Data should be categorical.
- Observations should be independent.
- Categories should be mutually exclusive.
- Expected frequencies should generally be sufficiently large.
### When to Use

Use Chi-Square when:

- Working with categorical data.
- Comparing observed and expected frequencies.
- Testing whether two categorical variables are associated.
### When Not to Use

Do not use Chi-Square when:

- Data is continuous without being categorized appropriately.
- Observations are dependent.
- Expected frequencies are too small for the approximation to be reliable.
### Simple Example

A company wants to know whether customer preference for four products is equally distributed.

Observed:

- Product A = 20
- Product B = 30
- Product C = 25
- Product D = 25

The Chi-Square Goodness-of-Fit Test can determine whether the observed distribution differs significantly from an expected distribution.

### Python Example – Goodness of Fit

```python
from scipy.stats import chisquare

observed = [20, 30, 25, 25]

chi_statistic, p_value = chisquare(observed)

print("Chi-Square Statistic:", chi_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

data = pd.DataFrame({
    "Purchased": ["Yes", "Yes", "No", "No", "Yes", "No", "Yes", "No"],
    "Gender": ["Male", "Female", "Male", "Female",
               "Male", "Female", "Female", "Male"]
})

contingency_table = pd.crosstab(data["Gender"], data["Purchased"])

chi_statistic, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table
)

print("Contingency Table:")
print(contingency_table)

print("Chi-Square Statistic:", chi_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Variables are associated")
else:
    print("No significant association found")
```

### Business Example

A company wants to determine whether customer gender and product purchase preference are related.

### AI/ML Example

Chi-Square can be used to analyze relationships between categorical features and categorical target variables.

It can also be used for categorical feature selection.

### Interpretation

If:

p-value ≤ 0.05

there is statistically significant evidence of an association or difference from the expected distribution.

If:

p-value > 0.05

there is not enough evidence to conclude that a significant association exists.

## 4. ANOVA
### Definition

ANOVA stands for Analysis of Variance.

It is used to determine whether there is a statistically significant difference between the means of three or more groups.

Instead of performing many T-Tests, ANOVA allows us to compare multiple groups using one overall test.

### Formula

ANOVA uses the F-statistic:

F = Between-Group Variation / Within-Group Variation

F = \frac{\text{between-group variation}}{\text{within-group variation}}

A large F-statistic means that the variation between groups is large compared with the variation within groups.

### Hypotheses

- H₀: All group means are equal.

- H₁: At least one group mean is different.

### Assumptions
- Dependent variable should be numerical.
- Groups should be independent.
- Observations should be independent.
- Data within each group should be approximately normally distributed.
- Variances should be reasonably similar across groups.
### When to Use

Use ANOVA when:

- Comparing three or more group means.
- The dependent variable is numerical.
- Groups are independent.
- ANOVA assumptions are reasonably satisfied.
### When Not to Use

Do not use standard one-way ANOVA when:

- Data is strongly non-normal with small samples.
- Groups are dependent when an independent-groups ANOVA is being considered.
- Variances are very different.
- There are only two groups and a simple T-Test is sufficient.

For strongly non-normal data, a non-parametric alternative such as Kruskal-Wallis may be considered.

### Simple Example

A company wants to compare the average sales generated by employees using three different training programs:

- Training A
- Training B
- Training C

ANOVA can determine whether the average sales differ significantly across the three groups.

### Python Example

```python
from scipy.stats import f_oneway

training_A = [70, 72, 75, 68, 74]
training_B = [80, 82, 85, 79, 84]
training_C = [65, 67, 70, 66, 69]

f_statistic, p_value = f_oneway(
    training_A,
    training_B,
    training_C
)

print("F-Statistic:", f_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
    print("At least one group mean is significantly different.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("No significant difference was detected between the group means.")
```

### Business Example

A company wants to compare sales performance across three different branches.

ANOVA can determine whether the average sales are significantly different.

### AI/ML Example

ANOVA can be used to investigate whether a numerical feature differs significantly across multiple categories or groups.

It can also help during feature analysis.

### Interpretation

If:

p-value ≤ 0.05

we reject H₀ and conclude that at least one group mean differs.

Important:

ANOVA tells us that at least one group is different, but it does not tell us exactly which groups are different.

Post-hoc tests such as Tukey's HSD can be used to identify the specific group differences.

## 5. Mann-Whitney U Test
### Definition

The Mann-Whitney U Test is a non-parametric statistical test used to compare two independent groups.

It is an alternative to the independent two-sample T-Test when the assumptions of the T-Test are not appropriate.

The test works using the ranks of observations rather than directly comparing their means.

### Assumptions
- Data should be at least ordinal.
- The two groups should be independent.
- Observations within each group should be independent.
- The test is commonly used when the data is not normally distributed.
### When to Use

Use the Mann-Whitney U Test when:

- Comparing two independent groups.
- Data is ordinal or continuous.
- Data is not normally distributed.
- There are outliers that make a parametric test less suitable.
### When Not to Use

Do not use it when:

- Groups are paired or dependent.
- There are more than two independent groups.
- Observations are not independent.
- The research question specifically requires comparing means.

For paired data, the Wilcoxon signed-rank test may be more appropriate.

### Simple Example

A company wants to compare customer satisfaction scores between:

- Company A
- Company B

Suppose the satisfaction scores are not normally distributed.

The Mann-Whitney U Test can be used to compare the two independent groups.

## Python Example

```python
from scipy.stats import mannwhitneyu

company_A = [70, 72, 75, 80, 68, 74]
company_B = [60, 65, 63, 70, 62, 67]

u_statistic, p_value = mannwhitneyu(
    company_A,
    company_B,
    alternative="two-sided"
)

print("U-Statistic:", u_statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
    print("There is a significant difference between the groups.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("No significant difference was detected.")
```

### Business Example

An online company wants to compare customer satisfaction ratings between two different website designs.

If the ratings are not normally distributed, the Mann-Whitney U Test can be used.

### AI/ML Example

It can be used to compare the distributions of a numerical feature between two independent groups when normality assumptions are not satisfied.

### Interpretation

If:

p-value ≤ 0.05

there is statistically significant evidence that the two groups differ.

If:

p-value > 0.05

there is not enough evidence to conclude that the groups differ.

## 6. Shapiro-Wilk Test
### Definition

The Shapiro-Wilk Test is a statistical test used to determine whether a sample is consistent with a normal distribution.

It is commonly used to check the normality assumption before applying parametric statistical tests.

Hypotheses

H₀: The data follows a normal distribution.

H₁: The data does not follow a normal distribution.

### Assumptions
- Data should be numerical.
- Observations should be independent.
- The test is intended for checking normality of a sample.
### When to Use

Use the Shapiro-Wilk Test when:

- You want to check whether data is approximately normally distributed.
- You are considering a parametric test such as a T-Test or ANOVA.
- The sample size is suitable for the test.
### When Not to Use

Do not rely only on the Shapiro-Wilk Test when:

- The sample is extremely large, because very small deviations from normality can become statistically significant.
- The data contains strong dependence between observations.
- You want to determine whether a variable is practically normal without also checking plots and domain knowledge.

It is better to use the test together with visual methods such as histograms and Q-Q plots.

### Simple Example

Suppose we have employee salaries and want to know whether they are approximately normally distributed before using a parametric test.

We can use the Shapiro-Wilk Test.

Python Example
from scipy.stats import shapiro

data = [48, 50, 52, 49, 51, 53, 50, 52, 49, 51]

statistic, p_value = shapiro(data)

print("Shapiro-Wilk Statistic:", statistic)
print("p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
    print("The data is not normally distributed.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence to say the data is non-normal.")
Business Example

A company wants to compare average employee salaries between groups using a T-Test.

Before applying the test, it can check whether the salary data is approximately normally distributed.

AI/ML Example

Normality testing can be useful during exploratory data analysis and statistical feature analysis before applying statistical methods that assume normality.

### Interpretation

If:

p-value > 0.05

we fail to reject H₀, so there is not enough evidence to conclude that the data is non-normal.

If:

p-value ≤ 0.05

we reject H₀, indicating evidence that the data deviates from normality.
