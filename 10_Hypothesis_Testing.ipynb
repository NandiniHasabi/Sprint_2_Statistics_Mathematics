# Hypothesis Testing

## 1. Null Hypothesis
Definition

The Null Hypothesis (H₀) is a statement that there is no significant difference, relationship, or effect in the data.

It is the default assumption that we test using statistical methods.

In simple words:

Null Hypothesis means "nothing has changed" or "there is no effect."

### Example

Suppose a company claims that the average delivery time is 30 minutes.

H₀: Average delivery time = 30 minutes

We start by assuming this statement is true and then check whether the sample data provides enough evidence to reject it.

### Simple Example

A company introduces a new training program and wants to know whether it improves employee performance.

H₀: The training program does not improve performance.
H₁: The training program improves performance.
### Numerical Example

Suppose the average marks of students before training are 70.

We collect a sample after training.

H₀: μ = 70
H₁: μ ≠ 70

Here, μ represents the population mean.

### Python Implementation

```python
import numpy as np

marks = np.array([72, 68, 75, 71, 74])

sample_mean = np.mean(marks)

print("Sample Mean:", sample_mean)
```

### Business Example

A company wants to check whether a new advertising campaign changes average sales.

H₀: The advertising campaign does not change sales.
H₁: The advertising campaign changes sales.
### AI/ML Example

Suppose we develop a new ML model and want to check whether it performs differently from the existing model.

H₀: New model performance = Existing model performance.
H₁: New model performance ≠ Existing model performance.

### Interpretation

The null hypothesis is the starting assumption of a hypothesis test.

If the statistical evidence is strong enough, we reject H₀.

If the evidence is not strong enough, we fail to reject H₀.

### 2. Alternative Hypothesis
Definition

The Alternative Hypothesis (H₁ or Ha) is a statement that there is a significant difference, relationship, or effect.

It is the opposite of the null hypothesis.

In simple words:

Alternative Hypothesis means "something has changed" or "there is an effect."

### Simple Example

A company wants to know whether a new training program improves employee performance.

H₀: Training does not improve performance.
H₁: Training improves performance.
### Numerical Example

Suppose the average delivery time is expected to be 30 minutes.

We want to check whether the actual average is different.

H₀: μ = 30
H₁: μ ≠ 30
### Python Implementation

```python
import numpy as np

delivery_time = np.array([28, 31, 29, 35, 32])

sample_mean = np.mean(delivery_time)

print("Sample Mean:", sample_mean)
```

### Business Example

An e-commerce company introduces a new website design.

H₀: New design does not change sales.
H₁: New design changes sales.
### AI/ML Example

We compare two ML models.

H₀: Both models have the same performance.
H₁: The models have different performance.

### Interpretation

The alternative hypothesis represents the effect or difference we are trying to find evidence for.

## 3. p-value
### Definition

The p-value tells us how likely it is to observe the sample result, or something more extreme, assuming that the null hypothesis is true.

In simple words:

p-value tells us how strong the evidence is against the null hypothesis.

A smaller p-value means stronger evidence against H₀.

### Common Rule

Usually:

If p-value < 0.05 → Reject H₀
If p-value ≥ 0.05 → Fail to reject H₀

The value 0.05 is a commonly used significance level.

### Simple Example

Suppose we perform a test and get:

p-value = 0.02

Since:

```python
0.02 < 0.05
```

We reject the null hypothesis.

### Numerical Example

Suppose:

H₀: New training has no effect.
H₁: New training has an effect.
p-value = 0.03

Since 0.03 < 0.05, there is statistically significant evidence against H₀.

### Python Implementation

```python
from scipy.stats import ttest_1samp

data = [52, 51, 53, 54, 55, 52, 51, 53]

t_stat, p_value = ttest_1samp(data, 50)

print("T-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example

A company wants to know whether a new marketing strategy increases sales.

If the p-value is 0.01, there is strong evidence that the strategy has an effect.

### AI/ML Example

Suppose we compare two ML models.

A statistical test gives:

```python
p-value = 0.02
```

This suggests that the difference in model performance is statistically significant.

### Interpretation

A low p-value provides stronger evidence against H₀.

A high p-value means that the observed result is not unusual under H₀.

### Important Point

p-value does not tell us the probability that H₀ is true.

## 4. Confidence Interval
### Definition

A Confidence Interval (CI) is a range of values used to estimate an unknown population parameter.

A common confidence level is 95%.

### For example:

Average Salary = ₹50,000
95% Confidence Interval = ₹47,000 to ₹53,000

This gives us a range of plausible values for the population mean.

### Formula

For a mean, a simplified confidence interval can be written as:

```python
Confidence Interval = Sample Mean ± Margin of Error
```

For a 95% confidence interval using a large sample:

```python
CI = x̄ ± 1.96 × (σ / √n)
```

Where:

x̄ = Sample Mean
σ = Standard Deviation
n = Sample Size
### Simple Example

Suppose the average salary from a sample is ₹50,000.

The 95% confidence interval is:

```python
₹47,000 to ₹53,000
```

This means the population mean is estimated to lie within this range.

### Numerical Example

Suppose:

```python
Mean = 100
Standard Deviation = 10
Sample Size = 100
```

For a 95% confidence interval:

```python
CI = 100 ± 1.96 × (10 / √100)

CI = 100 ± 1.96

Lower Limit = 98.04
Upper Limit = 101.96
```

```python
import numpy as np
from scipy.stats import t

data = np.array([45, 48, 50, 52, 49, 51, 47, 53, 50, 48])

mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

confidence_level = 0.95
alpha = 1 - confidence_level

t_critical = t.ppf(1 - alpha / 2, df=n-1)

margin_of_error = t_critical * (std / np.sqrt(n))

lower = mean - margin_of_error
upper = mean + margin_of_error

print("Mean:", mean)
print("95% Confidence Interval:", (lower, upper))
```

### Business Example

A company estimates that the average customer spending is ₹2,000.

The 95% confidence interval is:

```python
₹1,850 to ₹2,150
```

The company can use this range for planning and decision-making.

### AI/ML Example

Confidence intervals can be used to estimate the uncertainty of model performance.

### For example:

```python
Model Accuracy = 90%
95% CI = 87% to 93%
```

### Interpretation

A confidence interval gives a range of plausible values rather than a single estimate.

A narrow interval usually means more precise estimation.

A wide interval means more uncertainty.

## 5. Significance Level
### Definition

The Significance Level (α) is the threshold used to decide whether a result is statistically significant.

The most commonly used significance level is:

```python
α = 0.05
```

This means we accept a 5% risk of rejecting a true null hypothesis.

### Common Significance Levels

```python
0.10 → 10%
0.05 → 5%
0.01 → 1%
```

### Simple Example

Suppose:

Significance Level = 0.05
p-value = 0.03

Since:

```python
0.03 < 0.05
```

We reject H₀.

### Numerical Example

Suppose:

```python
α = 0.05
p-value = 0.08
```

Since:

```python
0.08 > 0.05
```

We fail to reject H₀.

### Python Implementation


```python
alpha = 0.05
p_value = 0.03

if p_value < alpha:
    print("Statistically Significant")
else:
    print("Not Statistically Significant")
```

### Business Example

A company tests whether a new product design increases sales.

The company chooses:

```python
α = 0.05
```

This becomes the decision threshold.

### AI/ML Example

A statistical test can be used to determine whether a feature has a significant relationship with a target variable.

### Interpretation

The significance level is selected before performing the hypothesis test.

## 6. Type I Error
### Definition

A Type I Error occurs when we reject the null hypothesis even though it is actually true.

In simple words:

Type I Error = False Positive

### Simple Example

Suppose:

H₀: A medicine has no effect.

If we reject H₀ and conclude that the medicine works when it actually does not, this is a Type I error.

### Numerical Example

If:

α = 0.05

there is a maximum 5% significance threshold for Type I error under the testing procedure.

### Python Implementation

```python
alpha = 0.05

print("Significance Level:", alpha)
print("Type I Error is a false positive.")
```

### Business Example

A company concludes that a new advertisement increases sales when it actually has no real effect.

### AI/ML Example

A fraud detection system incorrectly identifies a legitimate transaction as fraud.

This is a false positive.

### Interpretation

Type I Error means we detect an effect that does not actually exist.

## 7. Type II Error
### Definition

A Type II Error occurs when we fail to reject the null hypothesis even though the alternative hypothesis is actually true.

In simple words:

Type II Error = False Negative

### Simple Example

Suppose:

H₀: A medicine has no effect.

If the medicine actually works but we fail to detect its effect, this is a Type II error.

### Numerical Example

Suppose a new ML model actually performs better than an old model.

If our statistical test fails to detect the improvement, we have made a Type II error.

### Python Implementation

```python
print("Type II Error is a false negative.")
print("It happens when a real effect is not detected.")
```

### Business Example

A company introduces a better marketing strategy, but the statistical test fails to detect the improvement.

### AI/ML Example

A fraud detection system fails to identify an actual fraudulent transaction.

This is a false negative.

### Interpretation

Type II Error means we fail to detect an effect that actually exists.

## 8. One-tailed Test
### Definition

A One-tailed Test is a hypothesis test where the alternative hypothesis specifies a direction.

The direction can be:

Greater than
Less than

### Examples:

H₀: μ ≤ 50

H₁: μ > 50

or:

H₀: μ ≥ 50

H₁: μ < 50

### Simple Example

A company wants to check whether a new training program increases employee productivity.

H₀: μ ≤ 50

H₁: μ > 50

Since we are only interested in an increase, this is a one-tailed test.

### Numerical Example

Suppose:

Claimed mean = 50

Sample mean = 55

We test whether the population mean is significantly greater than 50.

### Python Implementation

```python
from scipy.stats import ttest_1samp

data = [52, 54, 55, 56, 53, 57, 55, 54]

t_statistic, p_two_tailed = ttest_1samp(data, 50)

if t_statistic > 0:
    p_one_tailed = p_two_tailed / 2
else:
    p_one_tailed = 1 - (p_two_tailed / 2)

print("T-Statistic:", t_statistic)
print("One-tailed p-value:", p_one_tailed)

if p_one_tailed <= 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example

A company wants to know whether a new advertisement increases sales.

### AI/ML Example

We can test whether a new feature improves model accuracy compared with an existing approach.

### Interpretation

A one-tailed test is used when the direction of the expected effect is specified before testing.

## 9. Two-tailed Test
### Definition

A Two-tailed Test is a hypothesis test where the alternative hypothesis checks for a difference in either direction.

Example:

H₀: μ = 50

H₁: μ ≠ 50

The population mean could be either greater than or less than 50.

### Simple Example

A company wants to know whether a new process changes the average production time.

They are interested in both:

Increase in production time
Decrease in production time

Therefore, a two-tailed test is appropriate.

### Numerical Example

Suppose:

H₀: μ = 50

H₁: μ ≠ 50

If the sample mean is 55 or 45, both directions can provide evidence against H₀.

### Python Implementation

```python
from scipy.stats import ttest_1samp

data = [52, 54, 55, 56, 53, 57, 55, 54]

t_statistic, p_value = ttest_1samp(data, 50)

print("T-Statistic:", t_statistic)
print("Two-tailed p-value:", p_value)

if p_value <= 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### Business Example

A company wants to know whether a new pricing strategy changes average customer spending, either increasing or decreasing it.

### AI/ML Example

We can test whether a new model has a different average error compared with an existing model.

The new model could have either higher or lower error.

### Interpretation

A two-tailed test is used when we want to detect a difference in either direction.
