# Measures of Dispersion

Measures of Dispersion are statistical measures that tell us how spread out or scattered the values in a dataset are.

Measures of Central Tendency such as Mean, Median, and Mode tell us about the center of the data.

Measures of Dispersion tell us how much the values differ from each other and from the center.

The main measures of dispersion covered in this notebook are:

- Range
- Variance
- Standard Deviation
- Quartiles
- Percentiles
- Interquartile Range (IQR)
- Coefficient of Variation

Understanding dispersion is important because two datasets can have the same average but very different levels of variation.

# 1. Range

Range is the simplest measure of dispersion.

It tells us the difference between the largest value and the smallest value in a dataset.

A larger range means that the values are more spread out.

A smaller range means that the values are closer together.

## Formula

Range = Maximum Value - Minimum Value

## Numerical Example

Consider the dataset:

10, 20, 30, 40, 50

Maximum value = 50

Minimum value = 10

Range = 50 - 10

Range = 40

Therefore, the range is 40.

## Business Example

Suppose a company records the daily sales for five days:

₹10,000, ₹12,000, ₹15,000, ₹18,000, ₹20,000

Maximum sales = ₹20,000

Minimum sales = ₹10,000

Range = ₹20,000 - ₹10,000

Range = ₹10,000

This tells the company how much the daily sales vary between the lowest and highest values.

## AI/ML Example

Range can be used during exploratory data analysis to understand the spread of a feature.

For example, if the age of customers ranges from 18 to 70:

Range = 70 - 18 = 52

This gives a basic idea of the variation in customer age.

## Interpretation

The range of the dataset is 40.

This means the difference between the highest and lowest values is 40.

A limitation of range is that it only uses two values: the minimum and maximum. Therefore, it can be strongly affected by outliers.

```python
data = [10, 20, 30, 40, 50]

range_value = max(data) - min(data)

print("Data:", data)
print("Maximum Value:", max(data))
print("Minimum Value:", min(data))
print("Range:", range_value)
```

## Output Interpretation

The maximum value is 50 and the minimum value is 10.

Therefore:

Range = 50 - 10 = 40

The values in the dataset span across a range of 40 units.

# 2. Variance

Variance measures how far the values in a dataset are spread out from the mean.

It is calculated by finding the difference between each value and the mean, squaring those differences, and then finding their average.

A small variance means that the values are close to the mean.

A large variance means that the values are more spread out from the mean.

## Formula for Population Variance

Population Variance:

σ² = Σ(x - μ)² / N

Where:

- σ² = Population variance
- x = Individual value
- μ = Population mean
- N = Number of observations

## Formula for Sample Variance

Sample Variance:

s² = Σ(x - x̄)² / (n - 1)

Where:

- s² = Sample variance
- x = Individual value
- x̄ = Sample mean
- n = Number of observations

The sample variance uses n - 1 instead of n.

## Numerical Example

Consider:

2, 4, 6

### Step 1 – Calculate Mean

Mean = (2 + 4 + 6) / 3

Mean = 12 / 3

Mean = 4

### Step 2 – Calculate Differences from Mean

2 - 4 = -2

4 - 4 = 0

6 - 4 = 2

### Step 3 – Square the Differences

(-2)² = 4

0² = 0

2² = 4

### Step 4 – Calculate Population Variance

Variance = (4 + 0 + 4) / 3

Variance = 8 / 3

Variance ≈ 2.67

Therefore, the population variance is approximately 2.67.

## Business Example

A company can use variance to understand how much employee salaries vary from the average salary.

If salary variance is high, employee salaries are widely spread.

If salary variance is low, employee salaries are closer to the average.

## AI/ML Example

Variance is important in Machine Learning because it helps us understand the spread of numerical features.

Variance is also important in:

- Feature analysis
- Feature selection
- Principal Component Analysis (PCA)
- Understanding data distribution

For example, a feature with extremely low variance may provide very little useful information and can sometimes be removed during feature selection.

## Interpretation

Variance measures the average squared distance of values from the mean.

A higher variance means greater spread.

A lower variance means less spread.

One limitation is that variance is expressed in squared units.

For example, if the original data is measured in kilograms, variance is measured in kilograms squared.

```python
import numpy as np

data = [2, 4, 6]

population_variance = np.var(data)

print("Data:", data)
print("Population Variance:", population_variance)
```

```python
sample_variance = np.var(data, ddof=1)

print("Sample Variance:", sample_variance)
```

## Output Interpretation

Population variance is approximately:

2.67

Sample variance is:

4.0

The difference occurs because population variance divides by N, while sample variance divides by N - 1.

# 3. Standard Deviation

Standard deviation measures how much the values in a dataset typically differ from the mean.

It is the square root of variance.

A small standard deviation means that the values are close to the mean.

A large standard deviation means that the values are more spread out.

## Formula

Population Standard Deviation:

σ = √(Σ(x - μ)² / N)

Sample Standard Deviation:

s = √(Σ(x - x̄)² / (n - 1))

## Relationship with Variance

Standard Deviation = √Variance

For example:

Variance = 25

Standard Deviation = √25

Standard Deviation = 5

## Numerical Example

Consider:

2, 4, 6

Mean = 4

Population variance ≈ 2.67

Therefore:

Standard Deviation = √2.67

Standard Deviation ≈ 1.63

## Business Example

A company can use standard deviation to understand the variation in monthly sales.

Suppose the average monthly sales are ₹50,000.

If the standard deviation is ₹2,000, sales are relatively close to the average.

If the standard deviation is ₹20,000, sales vary significantly.

## AI/ML Example

Standard deviation is commonly used in Machine Learning for feature scaling and standardization.

Standardization transforms data so that it has approximately:

Mean = 0

Standard Deviation = 1

The formula for a standardized value is:

z = (x - mean) / standard deviation

This is useful for algorithms such as:

- Logistic Regression
- Linear Regression
- K-Nearest Neighbors
- Support Vector Machines
- Principal Component Analysis

## Interpretation

Standard deviation is easier to interpret than variance because it uses the same units as the original data.

For example, if the data is measured in kilograms, the standard deviation is also measured in kilograms.

```python
import numpy as np

data = [2, 4, 6]

standard_deviation = np.std(data)

print("Data:", data)
print("Population Standard Deviation:", standard_deviation)
```

```python
sample_standard_deviation = np.std(data, ddof=1)

print("Sample Standard Deviation:", sample_standard_deviation)
```

## Output Interpretation

The population standard deviation is approximately 1.63.

This means that the values typically vary around 1.63 units from the mean of 4.

# 4. Quartiles

## Definition

Quartiles divide an ordered dataset into four equal parts.

There are three main quartiles:

- Q1 – First Quartile
- Q2 – Second Quartile
- Q3 – Third Quartile

### Q1 – First Quartile

Q1 represents the 25th percentile.

Approximately 25% of the data is below Q1.

### Q2 – Second Quartile

Q2 represents the 50th percentile.

Q2 is also the median.

Approximately 50% of the data is below Q2.

### Q3 – Third Quartile

Q3 represents the 75th percentile.

Approximately 75% of the data is below Q3.

## Numerical Example

Consider the dataset:

10, 20, 30, 40, 50, 60, 70, 80

Using a percentile-based calculation:

Q1 represents approximately the 25% point.

Q2 represents the 50% point.

Q3 represents the 75% point.

Quartiles help us understand how the data is distributed.

## Business Example

A company can divide customer spending into quartiles.

For example:

- Q1 can represent lower-spending customers.
- Q2 represents the middle of the customer spending distribution.
- Q3 represents higher-spending customers.

This can help the company identify customer segments.

## AI/ML Example

Quartiles are commonly used during exploratory data analysis.

They help us:

- Understand feature distributions
- Detect outliers
- Calculate IQR
- Create box plots
- Understand skewed data

## Interpretation

Quartiles divide the data into four parts.

Q1, Q2, and Q3 provide information about the position of observations within the dataset.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80]

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

print("Data:", data)
print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

## Output Interpretation

Q1 represents the 25th percentile.

Q2 represents the 50th percentile and is the median.

Q3 represents the 75th percentile.

These values help us understand how the observations are distributed across the dataset.

# 5. Percentiles

A percentile indicates the position of a value relative to the rest of the dataset.

The pth percentile is a value below which approximately p percent of the observations fall.

For example:

- 25th percentile means approximately 25% of the data is below that value.
- 50th percentile is the median.
- 75th percentile means approximately 75% of the data is below that value.
- 90th percentile means approximately 90% of the data is below that value.

## Formula

The exact calculation of a percentile can depend on the method used.

A commonly taught position formula is:

Position = (P / 100) × (n + 1)

Where:

- P = Desired percentile
- n = Number of observations

Different software packages may use different interpolation methods.

## Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70, 80

The 50th percentile is the middle of the data.

Therefore:

50th Percentile = 45

The 25th and 75th percentiles can be calculated using a percentile function.

## Business Example

An e-commerce company can use the 90th percentile of delivery time.

Suppose the 90th percentile delivery time is 5 days.

This means approximately 90% of deliveries are completed within 5 days, according to the dataset and percentile calculation.

## AI/ML Example

Percentiles can be used in Machine Learning for:

- Understanding feature distributions
- Detecting extreme observations
- Data analysis
- Setting thresholds
- Understanding model performance metrics

For example, the 95th percentile of a feature can be used as a threshold for identifying unusually high values.

## Interpretation

Percentiles help us understand the relative position of values in a dataset.

They are useful for comparing an observation with the rest of the data.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80]

percentiles = [25, 50, 75, 90]

for p in percentiles:
    value = np.percentile(data, p)
    print(f"{p}th Percentile:", value)
```

## Output Interpretation

The percentile values tell us where a particular percentage of observations fall.

For example, the 50th percentile represents the median.

The 75th percentile represents the value below which approximately 75% of observations fall.

# 6. Interquartile Range (IQR)

Interquartile Range, or IQR, measures the spread of the middle 50% of a dataset.

It is calculated as the difference between the third quartile and the first quartile.

IQR is useful because it is not strongly affected by extreme values.

## Formula

IQR = Q3 - Q1

Where:

- Q1 = First Quartile
- Q3 = Third Quartile

## Numerical Example

Suppose:

Q1 = 20

Q3 = 60

Then:

IQR = Q3 - Q1

IQR = 60 - 20

IQR = 40

Therefore, the IQR is 40.

## Outlier Detection Using IQR

IQR is commonly used to identify outliers.

Lower Bound:

Q1 - 1.5 × IQR

Upper Bound:

Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are commonly considered potential outliers.

## Numerical Example for Outlier Detection

Suppose:

Q1 = 20

Q3 = 60

IQR = 40

Lower Bound:

20 - (1.5 × 40)

= 20 - 60

= -40

Upper Bound:

60 + (1.5 × 40)

= 60 + 60

= 120

Therefore, values below -40 or above 120 would be considered potential outliers using this rule.

## Business Example

A company can use IQR to detect unusually high or low transaction amounts.

For example, if most customer transactions are between ₹500 and ₹5,000 but a few transactions are extremely high, IQR can help identify those unusual transactions.

## AI/ML Example

IQR is widely used for outlier detection during data preprocessing.

Outliers can sometimes negatively affect Machine Learning models.

The IQR method can help identify these values before training a model.

## Interpretation

A larger IQR means the middle 50% of the data is more spread out.

A smaller IQR means the middle 50% of the data is more concentrated.

IQR is more robust to extreme values than the range.

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70, 80]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
data = [10, 20, 30, 40, 50, 60, 70, 80, 200]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Output Interpretation

The IQR method calculates the lower and upper boundaries.

Any value below the lower bound or above the upper bound is considered a potential outlier.

In this example, 200 is much larger than the other observations, so it can be identified as an outlier using the IQR method.

# 7. Coefficient of Variation

Coefficient of Variation, or CV, is a relative measure of dispersion.

It compares the standard deviation with the mean.

It is useful for comparing the variability of datasets that have different means or different units.

## Formula

Coefficient of Variation:

CV = (Standard Deviation / Mean) × 100

## Numerical Example

Suppose:

Mean = 50

Standard Deviation = 5

Then:

CV = (5 / 50) × 100

CV = 10%

Therefore:

Coefficient of Variation = 10%

## Business Example

Suppose two investment options have:

Investment A:
Mean return = 10%
Standard deviation = 2%

Investment B:
Mean return = 20%
Standard deviation = 5%

Coefficient of Variation for Investment A:

CV = (2 / 10) × 100

CV = 20%

Coefficient of Variation for Investment B:

CV = (5 / 20) × 100

CV = 25%

Investment A has a lower relative variability.

## AI/ML Example

Coefficient of Variation can be used to compare the relative variability of different numerical features or measurements.

For example, suppose two features have different scales.

Feature A:

Mean = 100
Standard deviation = 10

Feature B:

Mean = 10
Standard deviation = 2

Their standard deviations cannot be directly compared without considering their scales.

Coefficient of Variation provides a relative comparison.

## Interpretation

A lower CV means lower variability relative to the mean.

A higher CV means higher variability relative to the mean.

CV is especially useful when comparing datasets with different scales.

CV should be interpreted carefully when the mean is zero or very close to zero because the calculation becomes unstable or undefined.

```python
mean = 50
standard_deviation = 5

cv = (standard_deviation / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
print("Coefficient of Variation:", cv, "%")
```

## Output Interpretation

The coefficient of variation is:

10%

This means that the standard deviation is 10% of the mean.

A lower CV generally indicates lower relative variability.
