# 1. What are Outliers?

## Definition

An outlier is a data point that is significantly different from the other observations in a dataset.

An outlier can be unusually high or unusually low compared with the rest of the data.

## Simple Example

Consider the following values:

10, 12, 11, 13, 12, 14, 100

Most values are between 10 and 14.

The value 100 is very different from the other values.

Therefore, 100 is an outlier.

## Numerical Example

Data:

[10, 12, 11, 13, 12, 14, 100]

The value 100 is much larger than the other observations.

Therefore:

Outlier = 100

## Business Example

Suppose the average order value of a company is around ₹500.

Most orders are:

₹400, ₹450, ₹500, ₹550, ₹600

But one order is:

₹50,000

₹50,000 may be an outlier.

## AI/ML Example

Suppose a Machine Learning dataset contains customer ages:

20, 22, 25, 28, 30, 32, 150

Age 150 is an obvious outlier.

Outliers can affect:

- Mean
- Variance
- Standard deviation
- Machine Learning models

## Python Implementation

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 14, 100])

print("Data:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
```

## Interpretation

The value 100 is much larger than the other observations.

The mean is affected strongly by the outlier, while the median is less affected.

# 2. Types of Outliers

## Definition

Outliers can be classified into different types depending on why and how they occur.

## 1. Point Outlier

A single observation is significantly different from the other observations.

### Example

10, 12, 11, 13, 100

Here, 100 is a point outlier.

## 2. Contextual Outlier

A value may be normal in one situation but unusual in another situation.

### Example

A temperature of 40°C may be normal in summer but unusual during winter.

The context determines whether the value is an outlier.

## 3. Collective Outlier

A group of observations together behaves differently from the rest of the data.

### Example

Normal website traffic is 1,000 users per hour.

Suddenly, the website receives:

10,000 users

12,000 users

15,000 users

20,000 users

This group may represent an unusual event.

## Business Example

A bank may normally see transactions between ₹100 and ₹50,000.

A transaction of ₹10,00,000 may be a point outlier.

## AI/ML Example

In fraud detection, an unusual transaction or a group of unusual transactions can be considered outliers.

## Interpretation

The type of outlier depends on the data and the context.

Not every unusual observation should automatically be removed.

# 3. Detecting Outliers

## Definition

Outlier detection is the process of identifying observations that are significantly different from the rest of the dataset.

Common methods include:

- IQR Method
- Z-Score Method
- Box Plot
- Scatter Plot

## Simple Example

Data:

[10, 12, 13, 15, 14, 100]

The value 100 appears unusual.

We can use statistical methods to determine whether it is an outlier.

## Python Implementation

```python
import numpy as np

data = np.array([10, 12, 13, 15, 14, 100])

mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std)
```

## Business Example

A company can detect unusually high sales transactions to investigate possible errors or fraud.

## AI/ML Example

Outlier detection is used in:

- Fraud detection
- Anomaly detection
- Network security
- Sensor monitoring
- Data cleaning

## Interpretation

Different techniques can be used depending on the distribution and characteristics of the data.

For example:

- IQR is useful for detecting outliers using quartiles.
- Z-score is useful when data is approximately normally distributed.
- Box plots provide a visual method.
- Scatter plots help detect unusual points between two variables.

# 4. IQR Method

## Definition

IQR stands for Interquartile Range.

It measures the spread of the middle 50% of the data.

## Formula

IQR = Q3 - Q1

Where:

Q1 = First Quartile

Q3 = Third Quartile

## Outlier Boundaries

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Any value below the lower bound or above the upper bound is considered an outlier.

## Numerical Example

Consider:

Data = [10, 12, 13, 14, 15, 16, 18, 20, 100]

Suppose:

Q1 = 12

Q3 = 20

IQR = Q3 - Q1

IQR = 20 - 12

IQR = 8

Lower Bound:

= 12 - 1.5 × 8

= 0

Upper Bound:

= 20 + 1.5 × 8

= 32

Any value greater than 32 is an outlier.

Therefore:

100 is an outlier.

## Python Implementation

```python
import numpy as np

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Business Example

A company can use the IQR method to identify unusually high or low transaction amounts.

## AI/ML Example

The IQR method can be used during data preprocessing to identify extreme values before training a Machine Learning model.

## Interpretation

Values outside the lower and upper boundaries are treated as potential outliers.

# 5. Z-Score Method

## Definition

A Z-score tells us how many standard deviations a value is away from the mean.

## Formula

Z = (X - μ) / σ

Where:

X = Data value

μ = Mean

σ = Standard deviation

## Common Rule

A common rule is:

If |Z| > 3

the observation may be considered an outlier.

## Numerical Example

Suppose:

Mean = 50

Standard Deviation = 10

Value = 90

Z = (90 - 50) / 10

Z = 40 / 10

Z = 4

Since:

|4| > 3

The value 90 may be considered an outlier.

## Python Implementation

```python
import numpy as np
from scipy.stats import zscore

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 100])

z_scores = zscore(data)

outliers = data[np.abs(z_scores) > 3]

print("Z-Scores:", z_scores)
print("Outliers:", outliers)
```

## Business Example

A company can use Z-scores to detect unusually high transaction amounts.

## AI/ML Example

Z-score based outlier detection can be used during data preprocessing, especially when the data is approximately normally distributed.

## Interpretation

A Z-score tells us how far a value is from the mean in terms of standard deviations.

For example:

Z = 0 → Value is at the mean.

Z = +2 → Value is 2 standard deviations above the mean.

Z = -2 → Value is 2 standard deviations below the mean.

# 6. Box Plot

## Definition

A box plot is a statistical visualization used to show the distribution of numerical data.

It displays:

- Minimum
- Q1
- Median
- Q3
- Maximum
- Potential outliers

## Important Parts

Q1 = 25th percentile

Median = 50th percentile

Q3 = 75th percentile

IQR = Q3 - Q1

Potential outliers are commonly shown beyond:

Q1 - 1.5 × IQR

and

Q3 + 1.5 × IQR

## Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 100])

plt.boxplot(data)

plt.title("Box Plot for Outlier Detection")
plt.ylabel("Values")
plt.show()
```

## Business Example

A company can use box plots to identify unusual:

- Sales values
- Customer spending
- Delivery times
- Transaction amounts

## AI/ML Example

Box plots can be used during Exploratory Data Analysis to detect extreme values in numerical features.

## Interpretation

The box represents the middle 50% of the data.

Points outside the whiskers may be potential outliers.

# 7. Scatter Plot

## Definition

A scatter plot displays individual data points using two numerical variables.

It is useful for identifying:

- Relationships
- Patterns
- Clusters
- Outliers

## Simple Example

Suppose we have:

Study Hours = [1, 2, 3, 4, 5, 10]

Marks = [40, 45, 50, 55, 60, 95]

The point (10, 95) may be unusual compared with the other observations.

## Python Implementation

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 10]
marks = [40, 45, 50, 55, 60, 95]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Scatter Plot for Outlier Detection")
plt.show()
```
