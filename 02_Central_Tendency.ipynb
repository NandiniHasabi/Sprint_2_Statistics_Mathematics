# Measures of Central Tendency

Measures of Central Tendency are statistical measures used to find the central or typical value of a dataset.

The main measures of central tendency are:

- Mean
- Median
- Mode
- Weighted Mean
- Geometric Mean
- Harmonic Mean

These measures help us summarize a dataset using one representative value.

## 1. Mean

### Definition

Mean is the average value of a dataset.

It is calculated by adding all the values and dividing the total by the number of values.

The mean is also called the arithmetic mean.

### Formula

Mean = Sum of all values / Number of values

Mathematically:

Mean = Σx / n

Where:

- Σx = Sum of all values
- n = Number of observations

### Numerical Example

Consider the following marks:

60, 70, 80, 90, 100

First, add all the values:

60 + 70 + 80 + 90 + 100 = 400

Number of values = 5

Mean = 400 / 5

Mean = 80

Therefore, the mean mark is 80.

### Business Example

A company wants to know the average monthly sales.

Suppose the monthly sales are:

₹20,000, ₹30,000, ₹40,000, ₹50,000, ₹60,000

The company can calculate the mean sales to understand the average monthly sales performance.

### AI/ML Example

In Machine Learning, the mean can be used to understand the average value of a numerical feature.

For example, if a dataset contains customer ages:

20, 25, 30, 35, 40

The mean age is:

30

This gives us an idea about the typical age of customers.

### Interpretation

The mean represents the arithmetic average of the data.

If the mean salary of employees is ₹40,000, it means the average salary is ₹40,000.

However, the mean can be strongly affected by extreme values or outliers.

For example:

10, 20, 30, 40, 500

The value 500 is much larger than the other values, so it increases the mean significantly.

```python
import numpy as np

data = [60, 70, 80, 90, 100]

mean = np.mean(data)

print("Data:", data)
print("Mean:", mean)
```

### Output Interpretation

The output shows:

Mean = 80

This means the average value of the dataset is 80.

## 2. Median

### Definition

Median is the middle value of a dataset after arranging the values in ascending or descending order.

The median divides the dataset into two equal parts.

If the number of observations is odd, the middle value is the median.

If the number of observations is even, the median is calculated by taking the average of the two middle values.

### Formula

For an odd number of observations:

Median = Middle value

For an even number of observations:

Median = (Middle value 1 + Middle value 2) / 2

### Numerical Example – Odd Number of Values

Consider:

10, 20, 30, 40, 50

There are 5 values.

The middle value is 30.

Therefore:

Median = 30

### Numerical Example – Even Number of Values

Consider:

10, 20, 30, 40

There are 4 values.

The two middle values are 20 and 30.

Median = (20 + 30) / 2

Median = 25

### Business Example

Suppose a company wants to understand employee salaries.

Consider:

₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹2,00,000

The value ₹2,00,000 is much higher than the other salaries.

In this situation, the median gives a better idea of the typical salary because it is less affected by the extreme value.

### AI/ML Example

The median is useful in Machine Learning when numerical data contains outliers.

For example, if a dataset contains customer income values with a few extremely high incomes, the median can represent the typical income better than the mean.

Median is also commonly used for handling missing numerical values through median imputation.

### Interpretation

The median represents the middle point of the ordered dataset.

The median is less affected by extreme values compared with the mean.

Therefore, median is useful for skewed data and datasets containing outliers.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

median = np.median(data)

print("Data:", data)
print("Median:", median)
```

### Output Interpretation

For the first dataset:

10, 20, 30, 40, 50

The median is 30 because 30 is the middle value.

For the second dataset:

10, 20, 30, 40

The median is 25 because:

(20 + 30) / 2 = 25

## 3. Mode

### Definition

Mode is the value that occurs most frequently in a dataset.

A dataset can have:

- One mode – Unimodal
- Two modes – Bimodal
- More than two modes – Multimodal
- No mode – When all values occur with the same frequency

### Formula

There is no mathematical formula for mode.

The mode is found by identifying the value with the highest frequency.

### Numerical Example

Consider:

10, 20, 20, 30, 40, 20, 50

The value 20 occurs three times.

Therefore:

Mode = 20

### Business Example

An online shopping company wants to find the most commonly purchased product category.

Suppose the categories purchased are:

Mobile, Laptop, Mobile, Tablet, Mobile, Laptop

The mode is:

Mobile

This tells the company that Mobile is the most frequently purchased category.

### AI/ML Example

Mode is useful for categorical data.

For example, if a dataset contains:

Red, Blue, Blue, Green, Blue

The mode is Blue.

Mode can also be used for filling missing values in categorical columns.

For example, if the most common customer city is Bengaluru, missing city values may sometimes be replaced with Bengaluru, depending on the business context.

### Interpretation

The mode represents the most frequently occurring value.

It is especially useful for categorical data where calculating a mean may not make sense.

```python
from collections import Counter

data = [10, 20, 20, 30, 40, 20, 50]

frequency = Counter(data)

mode = frequency.most_common(1)[0][0]

print("Data:", data)
print("Mode:", mode)
```

### Output Interpretation

The value 20 appears three times, which is more frequent than any other value.

Therefore:

Mode = 20

```python
import pandas as pd

data = [10, 20, 20, 30, 40, 20, 50]

mode = pd.Series(data).mode()

print("Mode:", mode.tolist())
```

## 4. Weighted Mean

### Definition

Weighted Mean is an average where different values are given different levels of importance or weights.

In a normal mean, every value has equal importance.

In a weighted mean, some values contribute more to the final result than others.

### Formula

Weighted Mean = Σ(wx) / Σw

Where:

- x = Value
- w = Weight
- wx = Value multiplied by its weight
- Σw = Sum of all weights

### Numerical Example

Suppose a student's final grade is calculated using:

Assignment = 80 marks with 20% weight

Exam = 90 marks with 80% weight

Weighted Mean:

= (80 × 0.20) + (90 × 0.80)

= 16 + 72

= 88

Therefore:

Weighted Mean = 88

### Business Example

A company evaluates an employee using:

Performance = 80 with 60% weight

Attendance = 90 with 20% weight

Communication = 70 with 20% weight

Weighted score:

= (80 × 0.60) + (90 × 0.20) + (70 × 0.20)

= 48 + 18 + 14

= 80

Therefore, the employee's weighted score is 80.

### AI/ML Example

Weighted averages can be useful when different observations have different importance.

For example, in a dataset, some observations may have higher importance or represent different amounts of data.

Weighted calculations are also used in evaluation metrics and loss functions in some Machine Learning applications.

### Interpretation

The weighted mean gives more importance to values with larger weights.

If an exam has a higher weight than an assignment, the exam has a greater effect on the final result.

```python
import numpy as np

values = [80, 90]
weights = [0.20, 0.80]

weighted_mean = np.average(values, weights=weights)

print("Values:", values)
print("Weights:", weights)
print("Weighted Mean:", weighted_mean)
```

### Output Interpretation

The weighted mean is 88.

The exam has a weight of 80%, so the exam score has a greater effect on the final result than the assignment score.

## 5. Geometric Mean

### Definition

Geometric Mean is an average calculated by multiplying all positive values and then taking the nth root, where n is the number of values.

It is mainly useful for growth rates, ratios, percentages, and values that multiply together.

### Formula

Geometric Mean = (x₁ × x₂ × x₃ × ... × xₙ)^(1/n)

Where:

- x₁, x₂, ..., xₙ are the values
- n is the number of values

### Numerical Example

Consider:

4 and 9

Geometric Mean:

= √(4 × 9)

= √36

= 6

Therefore:

Geometric Mean = 6

### Another Example

Suppose an investment grows by factors of:

1.10 and 1.20

Geometric Mean growth factor:

= √(1.10 × 1.20)

= √1.32

≈ 1.149

This represents an average multiplicative growth factor of approximately 1.149.

### Business Example

Geometric mean is useful for calculating average growth over multiple periods.

For example, a company may have different annual growth rates over several years.

The geometric mean can provide a better average growth rate when growth compounds over time.

### AI/ML Example

Geometric mean can be useful when combining ratios, rates, or multiplicative changes.

It can also be useful when analyzing metrics where relative changes are more meaningful than absolute changes.

### Interpretation

The geometric mean is useful when values are multiplied together or represent growth factors.

It is generally used with positive values.

It is different from the arithmetic mean because it is based on multiplication rather than addition.

```python
import numpy as np
from scipy.stats import gmean

data = [4, 9]

geometric_mean = gmean(data)

print("Data:", data)
print("Geometric Mean:", geometric_mean)
```

### Output Interpretation

The geometric mean of 4 and 9 is:

6

This is because:

√(4 × 9) = √36 = 6

## 6. Harmonic Mean

### Definition

Harmonic Mean is a type of average that is useful for rates, ratios, and values such as speed.

It gives more importance to smaller values compared with the arithmetic mean.

### Formula

Harmonic Mean = n / Σ(1/x)

Where:

- n = Number of values
- x = Each value

### Numerical Example

Consider:

4 and 6

Harmonic Mean:

= 2 / (1/4 + 1/6)

First:

1/4 + 1/6

= 3/12 + 2/12

= 5/12

Therefore:

Harmonic Mean = 2 / (5/12)

= 2 × 12/5

= 24/5

= 4.8

Therefore:

Harmonic Mean = 4.8

### Business Example

Suppose a delivery vehicle travels equal distances at different speeds.

For example:

60 km/h and 40 km/h

When averaging rates over equal distances, the harmonic mean is more appropriate than the arithmetic mean.

### AI/ML Example

The harmonic mean is used in Machine Learning evaluation.

A common example is the F1-score.

F1-score is the harmonic mean of:

- Precision
- Recall

Formula:

F1 = 2 × (Precision × Recall) / (Precision + Recall)

The harmonic mean ensures that a very low precision or recall strongly affects the final F1-score.

### Interpretation

The harmonic mean is useful when working with rates and ratios.

It gives more importance to smaller values.

It is especially useful when a high value should not compensate too much for a low value.

```python
from scipy.stats import hmean

data = [4, 6]

harmonic_mean = hmean(data)

print("Data:", data)
print("Harmonic Mean:", harmonic_mean)
```

### Output Interpretation

The harmonic mean of 4 and 6 is:

4.8

The result is lower than the arithmetic mean:

Arithmetic Mean = 5

Harmonic Mean = 4.8

This happens because the harmonic mean gives greater influence to the smaller value.
