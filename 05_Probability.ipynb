# Notebook 5 – Probability

Probability is a branch of mathematics that measures the likelihood or chance of an event happening.

The value of probability is always between 0 and 1.

- Probability = 0 means the event is impossible.
- Probability = 1 means the event is certain.
- Probability between 0 and 1 represents different levels of possibility.

Probability is very important in Data Science, Machine Learning, Artificial Intelligence, and decision-making.

## Formula

Probability of an event A:

P(A) = Number of favorable outcomes / Total number of possible outcomes

## Numerical Example

Suppose we have a fair coin.

Possible outcomes are:

{Head, Tail}

There is 1 favorable outcome for getting Head and 2 total outcomes.

P(Head) = 1 / 2 = 0.5

Therefore, the probability of getting Head is 0.5 or 50%.

## Business Example

A company may estimate the probability that a customer will purchase a product.

For example:

Probability of purchase = 0.30

This means there is a 30% chance that a randomly selected customer will purchase the product.

## AI/ML Example

Probability is used in Machine Learning to:

- Predict classes
- Estimate risks
- Calculate uncertainty
- Make decisions
- Perform classification

For example, a spam detection model may predict:

Spam = 0.90
Not Spam = 0.10

The model considers the email to have a 90% probability of being spam.

## Interpretation

Probability helps us quantify uncertainty and estimate how likely an event is to occur.

## Basic probability example

```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability of getting Head:", probability)
```

# 4. Independent Events

Two events are independent when the occurrence of one event does not affect the probability of the other event.

## Formula

For independent events A and B:

P(A ∩ B) = P(A) × P(B)

Where:

- P(A ∩ B) = Probability that both A and B occur
- P(A) = Probability of A
- P(B) = Probability of B

## Numerical Example

Suppose we toss a coin and roll a die.

Event A = Getting Head

P(A) = 1/2

Event B = Getting a 6

P(B) = 1/6

The coin result does not affect the die result.

Therefore, they are independent.

P(A ∩ B) = 1/2 × 1/6

P(A ∩ B) = 1/12

## Business Example

A customer receiving a promotional email and the weather being sunny may be treated as independent events if one does not influence the other in the situation being modeled.

## AI/ML Example

Independence assumptions are used in probabilistic Machine Learning models.

For example, Naive Bayes assumes conditional independence between features given the class.

## Interpretation

Independent events do not influence each other.

The probability of both events happening is calculated by multiplying their individual probabilities.

```python
probability_head = 1 / 2
probability_six = 1 / 6

probability_both = probability_head * probability_six

print("Probability of Head and 6:", probability_both)
```

# 5. Dependent Events

Two events are dependent when the occurrence of one event affects the probability of the other event.

## Formula

For dependent events:

P(A ∩ B) = P(A) × P(B|A)

Where:

- P(A) = Probability of event A
- P(B|A) = Probability of B given A has already occurred

## Numerical Example

Suppose a bag contains:

3 red balls
2 blue balls

Total = 5 balls

We select two balls without replacement.

Event A = First ball is red.

P(A) = 3/5

After selecting one red ball, only 4 balls remain.

Now there are 2 red balls remaining.

P(B|A) = 2/4

Therefore:

P(A ∩ B) = 3/5 × 2/4

P(A ∩ B) = 3/10 = 0.3

## Business Example

If a customer has already purchased one product, their probability of purchasing another related product may change.

Therefore, the events can be dependent.

## AI/ML Example

Many real-world variables are related.

For example:

Customer age and purchasing behavior may influence each other in a particular dataset.

Understanding dependencies helps us build better predictive models.

## Interpretation

In dependent events, the probability of the second event changes because the first event has already occurred.

```python
probability_first_red = 3 / 5
probability_second_red_given_first = 2 / 4

probability_both_red = (
    probability_first_red *
    probability_second_red_given_first
)

print("Probability of selecting two red balls:", probability_both_red)
```

# 6. Conditional Probability

Conditional Probability is the probability of an event occurring given that another event has already occurred.

It is written as:

P(A|B)

and read as:

"Probability of A given B."

## Formula

P(A|B) = P(A ∩ B) / P(B)

Where:

- P(A|B) = Probability of A given B
- P(A ∩ B) = Probability that A and B both occur
- P(B) = Probability of B

## Numerical Example

Suppose a class has 100 students.

60 students study Python.

40 students study both Python and SQL.

We want to find the probability that a student studies SQL given that the student studies Python.

P(SQL|Python) = P(SQL ∩ Python) / P(Python)

P(SQL|Python) = 40/60

P(SQL|Python) = 0.667

Therefore, the probability is approximately 66.7%.

## Business Example

A company may calculate:

Probability that a customer purchases a product given that they clicked an advertisement.

## AI/ML Example

Conditional probability is fundamental to:

- Classification
- Bayesian methods
- Naive Bayes
- Probabilistic models

## Interpretation

Conditional probability allows us to update the probability of an event when we have additional information.

P(A|B) = \frac{P(A \cap B)}{P(B)}

```python
probability_python = 60 / 100
probability_python_and_sql = 40 / 100

conditional_probability = (
    probability_python_and_sql /
    probability_python
)

print("P(SQL | Python):", conditional_probability)
```

# 7. Bayes' Theorem

Bayes' Theorem is a mathematical method used to update the probability of an event when new evidence becomes available.

It connects:

- Prior Probability
- Likelihood
- Evidence
- Posterior Probability

## Formula

P(A|B) = P(B|A) × P(A) / P(B)

Where:

- P(A|B) = Posterior probability
- P(B|A) = Likelihood
- P(A) = Prior probability
- P(B) = Evidence probability

## Numerical Example

Suppose:

1% of people have a particular disease.

P(Disease) = 0.01

A test correctly identifies a person with the disease 90% of the time.

P(Positive|Disease) = 0.90

Suppose the probability of getting a positive test result is 0.05.

Using Bayes' Theorem:

P(Disease|Positive)

= P(Positive|Disease) × P(Disease) / P(Positive)

= 0.90 × 0.01 / 0.05

= 0.18

Therefore, the probability of having the disease given a positive test result is 18% under these assumed probabilities.

## Business Example

A company can use Bayes' Theorem to estimate:

Probability that a customer will purchase a product given that the customer clicked an advertisement.

## AI/ML Example

Bayes' Theorem is used in:

- Naive Bayes
- Spam detection
- Medical diagnosis
- Risk prediction
- Classification
- Probabilistic reasoning

## Interpretation

Bayes' Theorem helps us update our belief about an event using new evidence.

It is especially useful when we have prior information and new observations.

P(A|B) = \frac{P(B|A)P(A)}{P(B)}

## Bayes' Theorem example

```python
p_disease = 0.01
p_positive_given_disease = 0.90
p_positive = 0.05

p_disease_given_positive = (
    p_positive_given_disease * p_disease
) / p_positive

print("P(Disease | Positive):", p_disease_given_positive)
```

# 8. Probability Rules

Probability follows several important rules.

## Rule 1 – Probability Range

The probability of an event is always between 0 and 1.

0 ≤ P(A) ≤ 1

## Rule 2 – Complement Rule

The probability that an event does not happen is:

P(Aᶜ) = 1 - P(A)

### Example

If:

P(A) = 0.7

Then:

P(Aᶜ) = 1 - 0.7

P(Aᶜ) = 0.3

## Rule 3 – Addition Rule

For two events:

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

This is used to calculate the probability that A or B occurs.

## Rule 4 – Multiplication Rule

For independent events:

P(A ∩ B) = P(A) × P(B)

## Rule 5 – Probability of Certain Event

P(S) = 1

The probability of the entire sample space is 1.

## Rule 6 – Probability of Impossible Event

P(∅) = 0

An impossible event has probability 0.

## Business Example

A company can use probability rules to calculate:

- Customer purchase probability
- Customer churn probability
- Fraud probability
- Product demand probability

## AI/ML Example

Probability rules are used in classification and probabilistic Machine Learning algorithms.

## Interpretation

Probability rules provide the mathematical foundation for calculating and combining probabilities.

## Complement Rule

```python
probability_purchase = 0.7

probability_no_purchase = 1 - probability_purchase

print("Probability of purchase:", probability_purchase)
print("Probability of no purchase:", probability_no_purchase)
```

# 10. Expected Value

Expected Value is the weighted average of all possible values of a random variable, where each value is weighted by its probability.

It represents the long-run average outcome we would expect if the experiment were repeated many times.

## Formula

For a discrete random variable:

E(X) = Σ xP(X = x)

Where:

- x = Possible value
- P(X = x) = Probability of that value
- E(X) = Expected value

## Numerical Example

Suppose a game has the following outcomes:

Win ₹100 with probability 0.5

Win ₹50 with probability 0.3

Win ₹0 with probability 0.2

Expected Value:

E(X) = (100 × 0.5) + (50 × 0.3) + (0 × 0.2)

E(X) = 50 + 15 + 0

E(X) = ₹65

Therefore, the expected value is ₹65.

## Business Example

A company can calculate the expected revenue from different customer outcomes.

For example, different products may have different sales probabilities and revenues.

## AI/ML Example

Expected value is used in:

- Decision making
- Reinforcement Learning
- Risk analysis
- Probabilistic modeling
- Expected loss calculations

## Interpretation

Expected value represents the average result we would expect over many repeated trials.

It does not necessarily mean that the expected value will be one of the actual outcomes.

\mathbb{E}[X] = \sum_x xP(X=x)

```python
# Expected Value example

outcomes = [100, 50, 0]
probabilities = [0.5, 0.3, 0.2]

expected_value = sum(
    outcome * probability
    for outcome, probability in zip(outcomes, probabilities)
)

print("Expected Value:", expected_value)
```
