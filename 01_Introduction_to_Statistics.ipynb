# Notebook 1 – Introduction to Statistics

## 1. What is Statistics?

Statistics is the science of collecting, organizing, analyzing, interpreting, and presenting data.

It helps us understand data, find patterns, make comparisons, and support better decisions.


### Example

Suppose a company has the following monthly sales:

10000, 12000, 15000, 13000, 18000

Using statistics, we can calculate the average sales and understand how sales are changing

### Business Example

A company can use statistics to analyze:
- Sales
- Customer feedback
- Employee performance
- Product demand
- Revenue
- Customer behavior

### AI/ML Example

In Artificial Intelligence and Machine Learning, statistics is used to understand datasets before building models.

For example, we can use statistics to:
- Find the average value of a feature
- Detect outliers
- Understand data distribution
- Find relationships between features
- Handle missing values
- Evaluate machine learning models

## 2. Why Statistics is Important in AI/ML

Statistics is important in AI/ML because machine learning models learn from data.

Before training a model, we need to understand the data properly.

Statistics helps us:

- Understand the dataset
- Summarize data
- Find patterns
- Detect outliers
- Handle missing values
- Understand relationships between variables
- Select useful features
- Understand probability
- Evaluate model performance
- Make predictions and decisions


### Real-World Example

Suppose we want to build a model that predicts house prices.

The dataset may contain:

- House area
- Number of bedrooms
- Location
- Age of house
- Price

Statistics helps us understand the distribution of these features and their relationships with house price.

### AI/ML Example

Before training a house-price prediction model, we can calculate:

- Mean house price
- Median house price
- Standard deviation
- Minimum and maximum values
- Correlation between area and price

This helps us understand the data before applying Machine Learning.

## 3. Statistics vs Machine Learning

Statistics and Machine Learning are closely related, but their main goals are different.

### Statistics

Statistics mainly focuses on:

- Understanding data
- Finding relationships
- Measuring uncertainty
- Making conclusions about populations
- Testing hypotheses
- Explaining patterns

### Machine Learning

Machine Learning mainly focuses on:

- Learning patterns from data
- Making predictions
- Classification
- Regression
- Clustering
- Automating decisions

### Difference

Statistics focuses more on:

**"What does the data tell us?"**

Machine Learning focuses more on:

**"What can we learn from the data to make predictions?"**

### Example

Suppose we have customer data.

Statistics can help us understand the average customer age and relationship between age and spending.

Machine Learning can use those features to predict how much a new customer may spend.


## 4. Descriptive Statistics

Descriptive statistics is used to summarize and describe the data we already have.

It does not try to make conclusions about a larger population.

Common descriptive statistics include:

- Mean
- Median
- Mode
- Range
- Variance
- Standard deviation
- Minimum
- Maximum
- Quartiles

### Example

Consider the following marks:

```text
60, 70, 80, 90, 100

### The mean is:

(60 + 70 + 80 + 90 + 100) / 5 = 80

So, the average mark is 80.

### Business Example

A company can calculate the average monthly sales to understand its sales performance.

### AI/ML Example

We can use descriptive statistics to understand the distribution of numerical features before training a machine learning model.

## 5. Inferential Statistics

Inferential statistics is used to make conclusions or predictions about a larger population using a sample.

It uses concepts such as:

Sampling
Probability
Confidence intervals
Hypothesis testing
Statistical tests

### Example

Suppose a company has 100,000 customers.

Instead of asking all 100,000 customers about their satisfaction, the company selects 1,000 customers.

The results from those 1,000 customers can be used to estimate the satisfaction of the complete customer population.

### Business Example

A company can survey a sample of customers and use the results to estimate overall customer satisfaction.

### AI/ML Example

Inferential statistics can help determine whether an observed difference in model performance or data groups is statistically meaningful.

## 6. Population

A population is the complete group of people, objects, or observations that we are interested in studying.

### Example

If we want to study the salaries of all employees in a company, all employees are the population.

### Business Example

All customers of an online shopping company are the population when studying customer behavior.

### AI/ML Example

All available records from which we want to understand a problem can be considered the population.

## 7. Sample

A sample is a smaller group selected from a population for analysis.

### Example

Suppose a company has 10,000 customers.

If we select 500 customers for a survey, those 500 customers are the sample.

### Business Example

A company can select 1,000 customers from its customer database to conduct a satisfaction survey.

### AI/ML Example

A sample can be used when the complete dataset is too large or expensive to process.

## 8. Census

A census is the process of collecting information from every member of a population.

### Example

If a company has 10,000 employees and collects information from all 10,000 employees, it is a census.

### Real-World Example

A national population census collects information about the population of a country.

### Difference Between Census and Sample
Census|	Sample
- Uses the entire population | Uses part of the population
- More expensive  | Less expensive
- Takes more time |	Usually faster
- Collects information from everyone | Collects information from selected members

## 9. Sampling

Sampling is the process of selecting a smaller group from a population for analysis.

Sampling is useful when studying the complete population is difficult, expensive, or time-consuming.

### Example

Population = 100,000 customers

Sample = 1,000 customers

The 1,000 customers are selected to represent the larger population.

### Business Example

A company can select a sample of customers to understand customer satisfaction.

### AI/ML Example

Sampling can be used to create a smaller dataset for analysis or experimentation.

### Important Point

A good sample should represent the population as closely as possible.

## 10. Variables

A variable is a characteristic or value that can change between observations.

### Examples
Age
Height
Salary
Temperature
Number of products
Customer rating
### Example

Consider this dataset:

Name	| Age | Salary
- A	    |  22 | 25000
- B    	|  25 | 30000
- C	    |  30 | 40000

Here:

Age is a variable.
Salary is a variable.
Name is also a variable.
### AI/ML Example

Variables that are used as inputs for a machine learning model are commonly called features.

## 11. Features

A feature is an input variable used by a machine learning model to make predictions or decisions.

### Example

Suppose we want to predict house prices.

The features can be:

House area
Number of bedrooms
Location
Number of floors
Age of house

The target variable can be:

House price
### AI/ML Example

For a disease prediction model, features may include:

Age
Blood pressure
Temperature
Heart rate
Glucose level

The model uses these features to predict the target.

### Definition

Feature = Input used by a Machine Learning model.

## 12. Observation

An observation is a single record or data point in a dataset.

Usually, one row represents one observation.

### Example

Consider this dataset:

Age	Salary	Experience
- 22	25000	1
- 25	30000	3
- 30	40000	5

Each row represents one observation.

Therefore:

Row 1 = Observation 1
Row 2 = Observation 2
Row 3 = Observation 3
### AI/ML Example

In a customer dataset, one observation can represent one customer.

## 13. Dataset

A dataset is a collection of related data containing observations and variables.

Usually, a dataset is organized in rows and columns.

### Example
 Age	    |Salary	|  Experience
- 22	| 25000 | 1
- 25	| 30000	| 3
- 30	| 40000	| 5
- 35	| 50000	| 8

Here:

The complete table is the dataset.
Each row is an observation.
Each column is a variable or feature.
### Business Example

A company may have a customer dataset containing:

Customer ID
Age
Gender
Location
Purchase amount
### AI/ML Example

A machine learning model learns patterns from a dataset during training.

## 14. Quantitative Data

Quantitative data is numerical data that can be measured or counted.

It represents quantities or amounts.

### Examples
Age = 25
Salary = ₹40,000
Height = 165.5 cm
Temperature = 30°C
Number of products = 10
Types of Quantitative Data

Quantitative data can be:

Discrete
Continuous
### Business Example

A company can analyze the numerical sales amount of each customer.

### AI/ML Example

Numerical features such as age, income, temperature, and price can be used as input features for ML models.

## 15. Qualitative Data

Qualitative data represents categories, qualities, or descriptions rather than numerical measurements.

### Examples
Gender
City
Product category
Color
Department
Customer type
### Example
Customer | City       |	Product Type
- A      |	Bengaluru |	Mobile
- B	     |Chennai     | Laptop
- C	     |Hyderabad	  |Tablet

Here:

City is qualitative data.
Product Type is qualitative data.
### Business Example

A company can analyze which product category is most popular.

### AI/ML Example

Qualitative data can be converted into numerical representations using techniques such as:

Label Encoding
One-Hot Encoding

## 16. Continuous Data

Continuous data is numerical data that can take any value within a range.

It can contain decimal values.

### Examples
Height = 165.5 cm
Weight = 55.7 kg
Temperature = 36.8°C
Time = 10.52 seconds
### Example

A person's height can be:

- 165 cm
- 165.5 cm
- 165.55 cm
- 165.555 cm

There can be many possible values within a range.

### Business Example

A manufacturing company can measure the weight of products.

### AI/ML Example

Continuous features such as height, weight, temperature, and income are commonly used in machine learning models.

## 17. Discrete Data

Discrete data consists of countable values.

Usually, discrete data contains whole numbers.

### Examples
- Number of students = 50
- Number of products = 20
- Number of employees = 100
- Number of calls = 25

We cannot normally have:

- 10.5 students

because students are counted as whole units.

### Business Example

A company can count the number of products sold in a day.

### AI/ML Example

The number of purchases, number of clicks, or number of customer complaints can be discrete variables.

## 18. Structured Data

Structured data is data organized in a fixed format, usually rows and columns.

### Examples
- Excel files
- CSV files
- SQL tables
- Relational databases
### Example
Employee_ID	| Name	| Age	Salary
- 101	Ravi	|25	|30000
- 102	Priya	|28|	40000
- 103	Arun	|30|	45000

This is structured data because it follows a fixed table structure.

### Business Example

Employee databases and sales databases are examples of structured data.

### AI/ML Example

Structured data is commonly used for:

- Regression
- Classification
- Customer prediction
- Sales prediction
- Fraud detection

## 19. Unstructured Data

Unstructured data does not follow a fixed row-and-column format.

### Examples
- Images
- Videos
- Audio
- Emails
- Social media posts
- Documents
- Text files
### Business Example

Customer reviews written as free text are unstructured data.

### AI/ML Example

Unstructured data is commonly used in:

- Computer Vision
- Natural Language Processing
- Speech Recognition
- Generative AI
### Example

A customer review such as:

"The product quality is excellent and delivery was very fast."

is unstructured text data.

An image uploaded by a customer is also unstructured data.
