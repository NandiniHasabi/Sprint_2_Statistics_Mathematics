# Statistical Data Visualization

## 1. Histogram
### Definition

A Histogram is a graph used to show the distribution of numerical data.

It divides data into intervals called bins and shows how many values fall into each bin.

### Purpose
- To understand the distribution of numerical data.
- To identify the center and spread of data.
- To identify skewness and possible outliers.
### When to Use

Use a histogram when:

- The data is numerical.
- We want to understand the distribution.
- We want to check whether data is approximately normal.
### Advantages
- Easy to understand.
- Shows data distribution clearly.
- Helps identify skewness and outliers.
- Useful for large datasets.
### Limitations
- Results depend on the number of bins.
- Not suitable for categorical data.
- Different bin sizes can give different appearances.
### Python Example

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Marks")
plt.show()
```

### Interpretation

The histogram shows how the marks are distributed across different ranges.

Most students have marks between approximately 50 and 80.

## 2. Bar Chart
### Definition

A Bar Chart represents categorical data using rectangular bars.

The height or length of each bar represents the value of that category.

### Purpose
- To compare different categories.
- To show counts or values for categories.
### When to Use

Use a bar chart when:

- Data is categorical.
- We want to compare categories.
- We want to display counts or totals.
### Advantages
- Easy to read.
- Good for comparing categories.
- Clearly shows differences between groups.
### Limitations
- Not suitable for continuous numerical data.
- Too many categories can make it difficult to read.
- Does not show distribution of numerical values.
### Python Example

```python
import matplotlib.pyplot as plt

departments = ["HR", "IT", "Finance", "Sales"]
employees = [10, 25, 15, 20]

plt.bar(departments, employees)
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Employees by Department")
plt.show()
```

### Interpretation

The IT department has the highest number of employees, while HR has the lowest.

## 3. Line Chart
### Definition

A Line Chart connects data points using lines.

It is mainly used to show trends or changes over time.

### Purpose
- To identify trends.
- To compare changes over time.
- To observe increases and decreases.
### When to Use

Use a line chart when:

- Data has a time component.
- We want to observe trends.
- Values change continuously.
### Advantages
- Clearly shows trends.
- Easy to identify increases and decreases.
- Useful for time-series data.
### Limitations
- Not suitable for unrelated categories.
- Too many lines can make the chart confusing.
- Can be misleading if the scale is inappropriate.
### Python Example

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 150, 140, 180]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```

### Interpretation

The IT department has the highest number of employees, while HR has the lowest.

## 4. Scatter Plot
### Definition

A Scatter Plot shows the relationship between two numerical variables using individual points.

### Purpose
- To identify relationships between variables.
- To identify positive or negative correlation.
- To detect possible outliers.
### When to Use

Use a scatter plot when:

- Both variables are numerical.
- We want to study their relationship.
- We want to visually check correlation.
### Advantages
- Clearly shows relationships.
- Helps identify correlation.
- Can identify outliers.
- Useful before building regression models.
### Limitations
- Not suitable for categorical data.
- Large datasets can contain overlapping points.
- Correlation does not prove causation.
### Python Example

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 60, 70, 80]

plt.scatter(study_hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

### Interpretation

The points show a positive relationship between study hours and marks. As study hours increase, marks generally increase.

## 5. Pie Chart
### Definition

A Pie Chart represents parts of a whole using circular slices.

The size of each slice represents the proportion of that category.

### Purpose
- To show percentage or proportion.
- To understand how a total is divided among categories.
### When to Use

Use a pie chart when:

- There are only a few categories.
- Categories represent parts of a whole.
- The total represents 100%.
### Advantages
- Easy to understand.
- Good for showing proportions.
- Visually attractive for simple datasets.
### Limitations
- Difficult to compare similar-sized slices.
- Not suitable for many categories.
- Not ideal for precise comparisons.
### Python Example

```python
import matplotlib.pyplot as plt

departments = ["IT", "HR", "Finance", "Sales"]
employees = [40, 20, 15, 25]

plt.pie(employees, labels=departments, autopct="%1.1f%%")
plt.title("Employee Distribution")
plt.show()
```

### Interpretation

The IT department represents the largest portion of employees, while Finance represents the smallest portion.

## 6. Density Plot
### Definition

A Density Plot shows the distribution of numerical data using a smooth curve.

It is also called a Kernel Density Estimate (KDE) plot.

### Purpose
- To understand the shape of data distribution.
- To identify peaks and spread.
- To compare distributions.
### When to Use

Use a density plot when:

- The data is numerical.
- We want a smooth representation of the distribution.
- We want to compare multiple distributions.
### Advantages
- Smooth representation of data.
- Easy to identify peaks.
- Useful for comparing distributions.
### Limitations
- The result depends on bandwidth.
- Can be difficult for beginners to understand.
- Not ideal for very small datasets.
### Python Example

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 80, 85, 90]

sns.kdeplot(marks, fill=True)
plt.xlabel("Marks")
plt.ylabel("Density")
plt.title("Density Plot of Marks")
plt.show()
```

### Interpretation

The highest part of the curve represents the range where the data values are concentrated.

## 7. Box Plot
### Definition

A Box Plot is used to summarize the distribution of numerical data.

It shows:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum
- Possible outliers
### Purpose
- To understand data spread.
- To identify the median.
- To detect outliers.
### When to Use

Use a box plot when:

- We want to compare distributions.
- We need to identify outliers.
- We want a summary of numerical data.
### Advantages
- Quickly shows the distribution.
- Clearly shows outliers.
- Useful for comparing groups.
- Compact representation.
### Limitations
- Does not show the exact distribution shape.
- Can be less intuitive for beginners.
- Individual data points are not shown.
### Python Example

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 80, 85, 100]

plt.boxplot(marks)
plt.ylabel("Marks")
plt.title("Box Plot of Marks")
plt.show()
```

### Interpretation

The box shows the middle 50% of the data. A point far away from the box may represent an outlier.

## 8. Violin Plot
### Definition

A Violin Plot combines a box plot with a density plot.

It shows both the summary statistics and the shape of the distribution.

### Purpose
- To understand distribution shape.
- To compare distributions between groups.
- To identify concentration of data.
### When to Use

Use a violin plot when:

- We want to compare multiple groups.
- We want to see both spread and distribution shape.
- The dataset has enough observations.
### Advantages
- Shows distribution shape.
- Shows median and spread.
- Useful for comparing groups.
### Limitations
- More difficult to understand than a box plot.
- Can be misleading with very small datasets.
- Requires more explanation for beginners.
### Python Example

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["IT", "IT", "IT", "IT", "HR", "HR", "HR", "HR"],
    "Salary": [40000, 45000, 50000, 55000, 30000, 35000, 38000, 40000]
}

sns.violinplot(x="Department", y="Salary", data=data)
plt.title("Salary Distribution by Department")
plt.show()
```

### Interpretation

The width of the violin shows where the data is more concentrated. A wider section means more values are present around that range.

## 9. Pair Plot
### Definition

A Pair Plot displays relationships between multiple numerical variables in a dataset.

It creates scatter plots between pairs of variables and distribution plots for individual variables.

### Purpose
- To understand relationships between multiple variables.
- To identify correlation.
- To identify patterns and clusters.
### When to Use

Use a pair plot when:

- The dataset contains multiple numerical features.
- We want to explore relationships quickly.
- We are performing Exploratory Data Analysis (EDA).
### Advantages
- Shows multiple relationships at once.
- Useful for EDA.
- Helps identify patterns and correlations.
### Limitations
- Can become crowded with many features.
- Requires more computation for large datasets.
- Difficult to interpret with many variables.
### Python Example

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Attendance": [60, 65, 70, 80, 90]
})

sns.pairplot(data)
plt.show()
```

### Interpretation

The pair plot helps us understand how Study Hours, Marks, and Attendance are related to each other.

## 10. Heatmap
### Definition

A Heatmap represents numerical values using different levels of shading.

It is commonly used to visualize a correlation matrix.

### Purpose
- To identify relationships between variables.
- To visualize correlation values.
- To quickly identify strongly related features.
### When to Use

Use a heatmap when:

- We have a correlation matrix.
- We want to visualize relationships between many variables.
- We want to identify highly correlated features.
### Advantages
- Easy to identify strong relationships.
- Can display many values in a compact format.
- Useful for feature analysis.
### Limitations
- Exact values may be difficult to compare without labels.
- Only shows relationships, not causation.
- Large matrices can become difficult to read.
### Python Example

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Attendance": [60, 65, 70, 80, 90],
    "Assignments": [2, 3, 4, 4, 5]
})

correlation_matrix = data.corr()

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

Values close to +1 indicate strong positive correlation.

Values close to -1 indicate strong negative correlation.

Values close to 0 indicate weak or no linear correlation.
