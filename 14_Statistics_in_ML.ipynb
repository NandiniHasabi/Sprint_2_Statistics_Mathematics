## 1. Data Cleaning

### Purpose
- Identify incorrect values.
- Find missing values.
- Detect outliers.
- Remove duplicate data.
- Understand the distribution of features.

We can use:

- Mean
- Median
- Mode
- Standard deviation
- Minimum and maximum
- Quartiles
- Percentiles

to understand the dataset.

### Simple Example

Suppose we have employee salaries:

```python
25000, 30000, 32000, 35000, 40000
```

We can calculate the mean salary to understand the average salary.

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35],
    "Salary": [25000, 30000, 35000, 40000, 45000]
})

print(data.describe())
```

- Count

- Minimum

- Maximum

## 2. Missing Value Handling
### Definition

Missing values are values that are not available in a dataset.

For example:

- Age
- 22
- 25
- NaN
- 30
- 35

Statistics can help us decide how to replace missing values.

### Common Methods
#### Mean

Use the mean when the data is approximately normally distributed and does not contain strong outliers.

#### Median

Use the median when the data is skewed or contains outliers.

#### Mode

Use the mode mainly for categorical data.

### Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [22, 25, None, 30, 35],
    "Salary": [25000, 30000, 35000, None, 45000]
})

data["Age"] = data["Age"].fillna(data["Age"].median())
data["Salary"] = data["Salary"].fillna(data["Salary"].mean())

print(data)
```

### Interpretation

The missing Age value is replaced using the median, while the missing Salary value is replaced using the mean.

### Business Example

A company may replace missing customer income values using the median income.

### AI/ML Example

Handling missing values prevents Machine Learning algorithms from failing or producing unreliable results.

## 3. Outlier Detection
### Definition

An outlier is a value that is significantly different from the other values in a dataset.

Example:

10, 12, 13, 14, 15, 100

Here, 100 may be an outlier.

### Statistical Methods

Common methods include:

- IQR Method
- Z-Score Method
- Box Plot
### IQR Formula
IQR = Q3 - Q1

Lower Bound:

Q1 - 1.5 × IQR

Upper Bound:

Q3 + 1.5 × IQR
### Python Example

```python
import numpy as np

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

### Interpretation

The code identifies values that fall outside the acceptable range.

### Business Example

In banking, an unusually large transaction may be detected as an outlier.

### AI/ML Example

Outliers can negatively affect models such as Linear Regression, so they should be investigated before training.

## 4. Feature Engineering
### Definition

Feature Engineering is the process of creating, transforming, or selecting features that help a Machine Learning model perform better.

Statistics helps us understand which transformations or new features may be useful.

### Examples

We can create:

- Age from Date of Birth
- Total Price from Quantity × Price
- Average Purchase from Total Purchase / Number of Purchases
- BMI from Height and Weight

```python
import pandas as pd

data = pd.DataFrame({
    "Quantity": [2, 5, 3, 10],
    "Price": [100, 50, 200, 30]
})

data["Total_Price"] = data["Quantity"] * data["Price"]

print(data)
```

### Interpretation

A new feature called Total_Price is created from two existing features.

### Business Example

An e-commerce company can create Total_Purchase_Value from quantity and product price.

### AI/ML Example

Useful engineered features can improve the predictive performance of Machine Learning models.

## 5. Feature Scaling
### Definition

Feature Scaling is the process of bringing numerical features to a similar scale.

This is important when features have very different ranges.

Example:

- Age       → 20 to 60
- Salary    → 20,000 to 100,000

Salary has a much larger numerical range.

### Why Scaling is Important

Some ML algorithms are sensitive to feature magnitude.

### Examples:

- K-Nearest Neighbors
- K-Means
- Support Vector Machine
- Logistic Regression
- Neural Networks
### Common Scaling Methods
- Standardization
- Min-Max Scaling
### Standardization Formula

z = (x - mean) / standard deviation

### Python Example

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [25000, 30000, 40000, 50000, 60000]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
```

### Interpretation

The features are transformed so that they have a similar scale.

### Business Example

Customer age and annual income can have very different ranges, so scaling can make them comparable for certain models.

## 6. Data Normalization
### Definition

Data Normalization transforms values into a specific range, commonly between 0 and 1.

### Min-Max Formula
X_scaled = (X - X_min) / (X_max - X_min)
### Python Example

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [25000, 30000, 40000, 50000, 60000]
})

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(data)

print(normalized_data)
```

### Interpretation

The values are transformed to a range between 0 and 1.

### Business Example

A company may normalize customer-related numerical features before applying a distance-based algorithm.

### AI/ML Example

Normalization is commonly useful for algorithms that depend on distances or gradient optimization.

## 7. Model Evaluation
### Definition

Model Evaluation is the process of measuring how well a Machine Learning model performs.

Statistics helps us determine whether the model's predictions are accurate and reliable.

### Common Evaluation Metrics
#### Regression
- Mean Absolute Error (MAE)
- Mean Squared Error (MSE)
- Root Mean Squared Error (RMSE)
- R² Score
#### Classification
- Accuracy
- Precision
- Recall
- F1 Score
- Confusion Matrix
### Python Example

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

actual = [100, 150, 200, 250, 300]
predicted = [110, 145, 190, 260, 295]

mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(actual, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

### Interpretation
- Lower MAE means smaller prediction errors.
- Lower MSE means smaller squared errors.
- Lower RMSE generally means better prediction accuracy.
- R² closer to 1 generally indicates a better fit.
### Business Example

A company can evaluate how accurately a model predicts sales.

### AI/ML Example

Model evaluation helps us compare different models and select a suitable one.

## 8. Linear Regression
### Definition

Linear Regression is a supervised Machine Learning algorithm used to predict a continuous numerical value.

It tries to find a relationship between input features and a target variable.

### Formula
y = b₀ + b₁x

Where:

- y = predicted value
- b₀ = intercept
- b₁ = coefficient
- x = input feature

Statistics is used to estimate the coefficients and measure how well the model fits the data.

### Python Example

```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([40, 50, 60, 70, 80])

model = LinearRegression()

model.fit(X, y)

prediction = model.predict([[6]])

print("Prediction:", prediction)
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
```

### Interpretation

The model learns the relationship between study hours and marks and predicts the marks for a new number of study hours.

### Business Example

Predicting house prices based on area.

### AI/ML Example

Predicting sales, salary, temperature, or house prices

## 9. Logistic Regression
### Definition

Logistic Regression is a supervised Machine Learning algorithm mainly used for classification problems.

It predicts the probability of a class.

For example:

0 → Not Fraud
1 → Fraud
### Statistical Concept

Logistic Regression uses probability and the sigmoid function.

The output probability is between 0 and 1.

### Python Example

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()

model.fit(X, y)

prediction = model.predict([[4]])
probability = model.predict_proba([[4]])

print("Prediction:", prediction)
print("Probability:", probability)
```

### Interpretation

The model predicts the class and also provides the probability of each class.

### Business Example

Predicting whether a customer will purchase a product.

### AI/ML Example
- Spam detection
- Fraud detection
- Disease classification
- Customer churn prediction

## 10. Naive Bayes
### Definition

Naive Bayes is a Machine Learning classification algorithm based on Bayes' Theorem.

It calculates the probability of a class based on the observed features.

### Bayes' Theorem
P(A|B) = P(B|A) × P(A) / P(B)
### Statistical Concept

Naive Bayes assumes that features are conditionally independent given the class.

### Python Example

```python
from sklearn.naive_bayes import GaussianNB

X = [
    [1, 20],
    [2, 21],
    [3, 22],
    [8, 30],
    [9, 31],
    [10, 32]
]

y = [0, 0, 0, 1, 1, 1]

model = GaussianNB()

model.fit(X, y)

prediction = model.predict([[4, 23]])

print("Prediction:", prediction)
```

### Interpretation

The model uses probability to determine which class is most likely for the new observation.

### Business Example

Classifying emails as spam or not spam.

### AI/ML Example

Naive Bayes is commonly used for:

- Text classification
- Spam detection
- Sentiment classification
- Document classification

## 11. Clustering
### Definition

Clustering is an unsupervised Machine Learning technique that groups similar data points together.

Statistics helps us understand the distribution and distance between observations.

### Example

Customers can be grouped into:

- Cluster 1 → Low spending customers
- Cluster 2 → Medium spending customers
- Cluster 3 → High spending customers
### K-Means Clustering

K-Means divides data into a specified number of clusters.

### Python Example

```python
import numpy as np
from sklearn.cluster import KMeans

data = np.array([
    [20, 20000],
    [22, 22000],
    [25, 25000],
    [40, 60000],
    [42, 65000],
    [45, 70000]
])

model = KMeans(n_clusters=2, random_state=42, n_init=10)

model.fit(data)

print("Cluster Labels:", model.labels_)
print("Cluster Centers:")
print(model.cluster_centers_)
```

### Interpretation

The algorithm groups observations based on similarity.

### Business Example

Customer segmentation based on age and spending.

### AI/ML Example

Clustering can be used for:

- Customer segmentation
- Image grouping
- Document grouping
- Anomaly detection

## 12. Principal Component Analysis (PCA)
### Definition

Principal Component Analysis (PCA) is a dimensionality reduction technique.

It transforms many correlated features into a smaller number of new features called Principal Components.

### Statistical Concept

PCA uses:

- Variance
- Covariance
- Eigenvalues
- Eigenvectors

The first principal component captures the maximum possible variance in the data.

### Purpose
- Reduce the number of features.
- Remove redundant information.
- Make models faster.
- Help visualize high-dimensional data.
### Python Example

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

data = pd.DataFrame({
    "Feature1": [10, 20, 30, 40, 50],
    "Feature2": [15, 25, 35, 45, 55],
    "Feature3": [12, 22, 32, 42, 52]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)

print("Original Shape:", data.shape)
print("PCA Shape:", pca_data.shape)
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```

### Interpretation

The original three features are reduced to two principal components while preserving as much information as possible.

### Business Example

A company may have hundreds of customer features. PCA can reduce the number of features before applying a Machine Learning model.

### AI/ML Example

PCA is commonly used for:

- Dimensionality reduction
- Data visualization
- Noise reduction
- Faster model training

## 13. Recommendation Systems
### Definition

A Recommendation System suggests products, movies, songs, or other items that a user may like.

Statistics helps recommendation systems understand user behavior and similarities between users or items.

### Statistical Concepts Used
- Mean ratings
- Similarity
- Correlation
- Probability
- User-item interactions
- Historical behavior
#### Example

Suppose users rate movies:

- User A → Movie 1: 5
- User A → Movie 2: 4
- User B → Movie 1: 5
- User B → Movie 2: 4

Because the users have similar rating patterns, the system may recommend similar movies to them.

### Python Example

```python
import pandas as pd

data = pd.DataFrame({
    "User": ["A", "A", "B", "B", "C", "C"],
    "Movie": ["Movie1", "Movie2", "Movie1", "Movie2", "Movie1", "Movie3"],
    "Rating": [5, 4, 5, 4, 3, 5]
})

user_movie_matrix = data.pivot_table(
    index="User",
    columns="Movie",
    values="Rating"
)

print(user_movie_matrix)

print("\nCorrelation:")
print(user_movie_matrix.corr())
```

### Interpretation

The user-item matrix shows how users rate different movies. Correlation can help identify movies with similar rating patterns.

### Business Example

E-commerce websites recommend products based on previous purchases and user behavior.

### AI/ML Example

Recommendation systems are used by:

- Netflix
- Amazon
- YouTube
- Spotify
- E-commerce platforms
