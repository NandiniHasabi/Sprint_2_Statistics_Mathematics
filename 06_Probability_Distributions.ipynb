# 1. Bernoulli Distribution

Bernoulli Distribution is a probability distribution for an experiment that has only two possible outcomes.

The two outcomes are usually represented as:

- 1 = Success
- 0 = Failure

A Bernoulli experiment is performed only once.

Examples:

- Head or Tail
- Pass or Fail
- Yes or No
- Success or Failure
- Spam or Not Spam
- Customer buys or does not buy

## Main Characteristics

- There are only two possible outcomes.
- The experiment has one trial.
- The probability of success is p.
- The probability of failure is 1 - p.

## Formula

The Bernoulli probability mass function is:

P(X = x) = p^x × (1 - p)^(1-x)

Where:

- X = Random variable
- x = 0 or 1
- p = Probability of success
- 1 - p = Probability of failure

## Numerical Example

Suppose a customer has a 70% probability of purchasing a product.

Probability of purchase:

p = 0.70

Probability of no purchase:

1 - p = 0.30

Therefore:

P(X = 1) = 0.70

P(X = 0) = 0.30

## Where It Is Used

Bernoulli Distribution is used when an experiment has only two possible outcomes.

Examples:

- Customer purchase
- Loan approval
- Email spam detection
- Medical test result
- Machine failure
- Website click

## Real-World Example

A company sends an advertisement to one customer.

The customer either:

1. Clicks the advertisement
2. Does not click the advertisement

This is a Bernoulli experiment because there are only two possible outcomes.

## AI/ML Example

In binary classification, the target variable can have two classes.

For example:

0 = Not Spam

1 = Spam

The outcome can be modeled as a Bernoulli random variable.

## Interpretation

Bernoulli Distribution is suitable when there is only one trial and exactly two possible outcomes.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)

plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.xticks([0, 1], ["Failure", "Success"])

plt.show()
```

## Visualization Interpretation

The graph shows two possible outcomes:

- 0 represents Failure
- 1 represents Success

Since the probability of success is 0.7, the probability of success is higher than the probability of failure.

Therefore:

P(Success) = 0.7

P(Failure) = 0.3

# 2. Binomial Distribution

Binomial Distribution describes the number of successes in a fixed number of independent Bernoulli trials.

Each trial must have two possible outcomes:

- Success
- Failure

The probability of success must remain the same for every trial.

## Main Characteristics

A binomial distribution requires:

- Fixed number of trials
- Two possible outcomes
- Independent trials
- Same probability of success for each trial

## Formula

The Binomial Probability Mass Function is:

P(X = k) = C(n,k) × p^k × (1-p)^(n-k)

Where:

- n = Number of trials
- k = Number of successes
- p = Probability of success
- 1-p = Probability of failure
- C(n,k) = Combination of n items taken k at a time

## Numerical Example

Suppose a customer has a 60% probability of purchasing a product.

We observe 5 customers.

We want to find the probability that exactly 3 customers purchase the product.

Here:

n = 5

k = 3

p = 0.6

The binomial formula can be used to calculate the probability.

## Where It Is Used

Binomial Distribution is used when we want to count the number of successes in a fixed number of trials.

Examples:

- Number of successful sales
- Number of defective products
- Number of customers who click an advertisement
- Number of successful predictions
- Number of patients responding to treatment

## Real-World Example

A company sends an advertisement to 10 customers.

Each customer either clicks or does not click the advertisement.

The company can use Binomial Distribution to estimate the probability of getting exactly 6 clicks.

## AI/ML Example

Binomial Distribution can be used to model the number of successful outcomes in repeated binary classification trials.

For example, we can model the number of correct predictions out of a fixed number of binary predictions when the assumptions of the binomial model are reasonable.

## Interpretation

Bernoulli describes one binary trial.

Binomial describes the number of successes across multiple independent Bernoulli trials.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.6

x = np.arange(0, n + 1)

probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)

plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")

plt.show()
```

## Visualization Interpretation

The graph shows the probability of getting different numbers of successes in 10 trials.

The highest probabilities occur around the expected number of successes.

Expected number of successes:

E(X) = n × p

E(X) = 10 × 0.6

E(X) = 6

Therefore, the distribution is centered around approximately 6 successes.

# 3. Poisson Distribution

Poisson Distribution is used to model the number of times an event occurs during a fixed interval of time or space.

The event should occur independently and at an approximately constant average rate.

Examples:

- Number of customer arrivals per hour
- Number of calls received per minute
- Number of website requests per second
- Number of accidents per month
- Number of machine failures per day

## Formula

The Poisson probability mass function is:

P(X = k) = (e^(-λ) × λ^k) / k!

Where:

- X = Number of events
- k = Number of events
- λ = Average number of events in the interval
- e = Euler's number
- k! = Factorial of k

## Numerical Example

Suppose a customer support center receives an average of 5 calls per hour.

Therefore:

λ = 5

Poisson Distribution can be used to calculate the probability of receiving exactly 3 calls in one hour.

## Where It Is Used

Poisson Distribution is used for counting events within a fixed interval.

Examples:

- Number of calls per hour
- Number of website visits per minute
- Number of machine failures per month
- Number of customers arriving at a store
- Number of accidents per day

## Real-World Example

A hospital receives an average of 8 emergency patients per hour.

Poisson Distribution can be used to estimate the probability of receiving a specific number of emergency patients during one hour.

## AI/ML Example

Poisson Distribution can be used in Machine Learning and data analysis for count-based data.

Examples include:

- Number of user interactions
- Number of events
- Number of failures
- Number of visits
- Number of occurrences

## Interpretation

Poisson Distribution is useful when we are counting how many times an event occurs during a fixed interval.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 5

x = np.arange(0, 16)

probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)

plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")

plt.show()
```

## Visualization Interpretation

The graph shows the probability of different numbers of events occurring when the average number of events is 5.

The distribution is centered around the average event rate.

Here:

λ = 5

This means the average number of events in the interval is 5.

# 4. Uniform Distribution

Uniform Distribution is a probability distribution where all values within a specified range have equal probability density.

There are two types:

- Discrete Uniform Distribution
- Continuous Uniform Distribution

In this notebook, we focus mainly on Continuous Uniform Distribution.

## Formula

For a continuous uniform distribution:

f(x) = 1 / (b - a)

Where:

- a = Lower limit
- b = Upper limit

## Numerical Example

Suppose a random number is selected between 10 and 20.

Every value within this interval has the same probability density.

Here:

a = 10

b = 20

## Where It Is Used

Uniform Distribution is used in:

- Random number generation
- Simulation
- Random sampling
- Testing
- Game development
- Computer simulations

## Real-World Example

Suppose a customer arrives at a store randomly between 10:00 AM and 11:00 AM.

If every time within this hour is equally likely, arrival time can be modeled using a uniform distribution.

## AI/ML Example

Uniform Distribution can be used for:

- Random initialization
- Simulation
- Data generation
- Hyperparameter sampling
- Testing Machine Learning algorithms

For example, a parameter can be randomly selected between a minimum and maximum value.

## Interpretation

Uniform Distribution means that values are spread relatively evenly across the specified range.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=30, density=True)

plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```

## Visualization Interpretation

The histogram is approximately flat across the range.

This indicates that the values are distributed relatively evenly between 0 and 10.

There is no strong concentration around a particular value.

# 5. Normal Distribution

Normal Distribution is a continuous probability distribution with a bell-shaped and symmetric curve.

It is one of the most important distributions in statistics.

In a perfectly normal distribution:

Mean = Median = Mode

## Formula

The probability density function of a normal distribution is:

f(x) = 1 / (σ√(2π)) × e^(-(x-μ)² / (2σ²))

Where:

- x = Data value
- μ = Mean
- σ = Standard deviation
- π = Mathematical constant
- e = Euler's number

## Main Characteristics

- Bell-shaped
- Symmetric
- Mean = Median = Mode
- Most observations are near the mean
- Fewer observations occur far from the mean

## Numerical Example

Suppose the average height of students is:

Mean = 165 cm

Standard Deviation = 5 cm

The heights can approximately follow a normal distribution.

Most students will have heights close to 165 cm.

## Where It Is Used

Normal Distribution is used in:

- Statistics
- Data analysis
- Statistical testing
- Measurement analysis
- Error modeling
- Machine Learning

## Real-World Example

Human height within a suitable population is often approximately normally distributed.

Most people have heights near the average, while very short and very tall values are less common.

## AI/ML Example

Normal Distribution is important in:

- Gaussian Naive Bayes
- Gaussian Mixture Models
- Statistical modeling
- Error analysis
- Feature analysis

## Interpretation

Normal Distribution describes data that is concentrated around the mean and becomes less frequent as we move away from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30, density=True)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```

## Visualization Interpretation

The histogram has an approximately bell-shaped curve.

Most values are located around the mean of 50.

The frequency decreases as we move farther away from the mean.

This represents the basic shape of a normal distribution.

# 6. Exponential Distribution

Exponential Distribution is a continuous probability distribution commonly used to model the waiting time between events.

It is closely related to the Poisson Distribution.

Poisson Distribution counts the number of events in an interval.

Exponential Distribution models the waiting time between events.

## Formula
 
The probability density function is:

f(x) = λe^(-λx)

for x ≥ 0

Where:

- x = Waiting time
- λ = Rate parameter
- e = Euler's number

## Numerical Example

Suppose a customer support center receives an average of 5 calls per hour.

The average rate is:

λ = 5 calls per hour

Exponential Distribution can be used to model the waiting time between two consecutive calls.

## Where It Is Used

Exponential Distribution is used for:

- Waiting times
- Time between customer arrivals
- Machine failure times
- Service times
- Network delays
- Queueing systems

## Real-World Example

Suppose customers arrive at a restaurant randomly.

We can use Exponential Distribution to model the time between two consecutive customer arrivals, assuming the arrival process satisfies the required assumptions.

## AI/ML Example

Exponential Distribution can be used in:

- Survival analysis
- Reliability analysis
- Event-time modeling
- Queueing analysis
- Time-to-event problems

For example, it can help model the time until a machine failure.

## Interpretation

Exponential Distribution is mainly used to model waiting time until the next event occurs.
