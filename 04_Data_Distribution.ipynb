# 1. Normal Distribution

Normal Distribution is a continuous probability distribution in which the data is symmetrically distributed around the mean.

It has a bell-shaped curve.

In a perfectly normal distribution:

Mean = Median = Mode

Most observations are located near the center, while fewer observations occur as we move away from the center.

## Main Characteristics

- Bell-shaped curve
- Symmetric around the mean
- Mean, median, and mode are equal
- Most values are close to the mean
- Fewer values occur far from the mean
- The two sides of the distribution are mirror images

## Formula

The probability density function of a normal distribution is:

f(x) = 1 / (σ√(2π)) × e^(-(x - μ)² / (2σ²))

Where:

- x = Data value
- μ = Mean
- σ = Standard deviation
- π = Mathematical constant
- e = Euler's number

The formula is mainly used to describe the mathematical shape of the distribution.

## Numerical Example

Suppose the heights of students have:

Mean = 165 cm

Standard Deviation = 5 cm

In a normal distribution, most students will have heights close to 165 cm.

Values such as 163 cm, 166 cm, and 168 cm are closer to the mean.

Very small or very large heights are less common.

## Business Example

A company may analyze the delivery times of products.

If delivery times approximately follow a normal distribution, most deliveries will happen close to the average delivery time.

## AI/ML Example

Normal distributions are important in Machine Learning and statistics.

They are used in:

- Data analysis
- Feature scaling
- Statistical testing
- Linear regression
- Gaussian-based models
- Understanding model errors

## Interpretation

A normal distribution shows that values are concentrated around the mean and become less frequent as they move away from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, density=True)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```

## Output Interpretation

The histogram forms an approximately bell-shaped curve.

Most values are located around the mean of 50.

As the values move farther away from 50, their frequency decreases.

This represents the general shape of a normal distribution.

# 2. Uniform Distribution

Uniform Distribution is a probability distribution in which all values within a specified range have equal probability.

In a continuous uniform distribution, every interval of the same length has the same probability.

The graph generally has a rectangular shape.

## Formula

For a continuous uniform distribution:

f(x) = 1 / (b - a)

Where:

- a = Minimum value
- b = Maximum value

For values between a and b.

## Numerical Example

Suppose a random number is selected between 1 and 10.

If the distribution is uniform, values across the range have equal probability density.

For example, there is no preference for numbers near the center.

## Business Example

A company may use a uniform distribution to simulate random arrival times within a fixed time interval.

For example, a system may generate random test times between 10:00 AM and 11:00 AM.

## AI/ML Example

Uniform distributions can be used for:

- Random data generation
- Simulation
- Random initialization
- Testing algorithms
- Hyperparameter sampling

For example, a Machine Learning algorithm can randomly select a parameter value from a specified range.

## Interpretation

Uniform Distribution means the values are evenly distributed across the selected range.

Unlike normal distribution, values near the center are not more common than values near the boundaries.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.uniform(low=0, high=10, size=1000)

plt.hist(data, bins=20, density=True)

plt.title("Uniform Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```

## Output Interpretation

The histogram is approximately flat across the range from 0 to 10.

This shows that the values are distributed relatively evenly across the range.

# 3. Gaussian Distribution

Gaussian Distribution is another name for the Normal Distribution.

It is named after the mathematician Carl Friedrich Gauss.

A Gaussian distribution has a bell-shaped and symmetric curve.

## Main Characteristics

- Bell-shaped
- Symmetric
- Mean = Median = Mode
- Controlled by mean and standard deviation
- Values close to the mean are more common
- Extreme values are less common

## Formula

The Gaussian probability density function is:

f(x) = 1 / (σ√(2π)) × e^(-(x - μ)² / (2σ²))

Where:

- μ = Mean
- σ = Standard deviation
- x = Data value

## Numerical Example

Suppose:

Mean = 100

Standard Deviation = 10

The distribution will be centered around 100.

Values close to 100 are more common than values such as 50 or 150.

## Business Example

A company can analyze employee performance scores.

If performance scores approximately follow a Gaussian distribution, most employees may have scores close to the average.

## AI/ML Example

Gaussian distributions are important in:

- Gaussian Naive Bayes
- Gaussian Mixture Models
- Statistical modeling
- Error modeling
- Feature analysis

## Interpretation

Gaussian Distribution and Normal Distribution refer to the same basic probability distribution.

The terms can be used interchangeably in most statistical contexts.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.normal(loc=100, scale=10, size=1000)

plt.hist(data, bins=30, density=True)

plt.title("Gaussian Distribution")
plt.xlabel("Values")
plt.ylabel("Density")

plt.show()
```

## Output Interpretation

The graph shows a bell-shaped distribution centered around approximately 100.

Most observations are close to the mean.

This is the characteristic shape of a Gaussian or Normal Distribution.

# 4. Left Skew

Left Skew is also called Negative Skewness.

A left-skewed distribution has a longer tail on the left side.

Most of the observations are concentrated toward the right side, while a smaller number of low values create the long left tail.

## Relationship

In a typical left-skewed distribution:

Mean < Median < Mode

The mean is pulled toward the lower values because of the long left tail.

## Numerical Example

Consider:

10, 70, 75, 80, 82, 85, 88

The value 10 is much smaller than the other values.

This creates a longer tail toward the left.

Therefore, the distribution is left-skewed.

## Business Example

Suppose most employees receive high performance scores, but a small number of employees receive very low scores.

The distribution of scores may become left-skewed.

## AI/ML Example

Skewness is important during Exploratory Data Analysis.

A highly skewed feature may affect statistical analysis and some Machine Learning models.

Data transformations such as logarithmic or other suitable transformations can sometimes be used to reduce skewness, depending on the data.

## Interpretation

In left skew:

- Tail extends toward the left
- Most values are toward the right
- Mean is generally less than the median
- A few low values pull the mean downward

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = -np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)

plt.title("Left-Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

## Output Interpretation

The distribution has a longer tail toward the left side.

Most observations are concentrated toward the higher-value side.

This represents a left-skewed distribution.

# 5. Right Skew

Right Skew is also called Positive Skewness.

A right-skewed distribution has a longer tail on the right side.

Most observations are concentrated toward the left side, while a smaller number of high values create the long right tail.

## Relationship

In a typical right-skewed distribution:

Mode < Median < Mean

The mean is pulled toward the higher values because of the long right tail.

## Numerical Example

Consider:

10, 20, 25, 30, 32, 35, 100

The value 100 is much larger than the other values.

This creates a long tail toward the right.

Therefore, the distribution is right-skewed.

## Business Example

Income data is often right-skewed because most people may have moderate incomes while a smaller number of people have very high incomes.

## AI/ML Example

Right-skewed features can occur in:

- Income
- House prices
- Transaction amounts
- Website visits
- Customer spending

Understanding skewness helps during data preprocessing and Exploratory Data Analysis.

## Interpretation

In right skew:

- Tail extends toward the right
- Most values are toward the left
- Mean is generally greater than the median
- A few high values pull the mean upward

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)

plt.title("Right-Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

## Output Interpretation

The distribution has a longer tail toward the right side.

Most observations are concentrated toward the lower-value side.

This represents a right-skewed distribution.

# 6. Symmetric Distribution

A symmetric distribution is a distribution where the left and right sides have approximately the same shape around the center.

If the distribution is perfectly symmetric, the values on both sides of the center are balanced.

Normal Distribution is a common example of a symmetric distribution.

## Main Characteristics

- Left and right sides are similar
- No major skewness
- The center represents the balance point
- In a perfectly normal distribution:

Mean = Median = Mode

## Numerical Example

Consider:

10, 20, 30, 40, 50

The data is centered around 30.

The values are balanced around the center:

10 and 50

20 and 40

30 is the center.

This dataset is symmetric.

## Business Example

Suppose employee performance scores are evenly distributed around the average.

There are approximately similar numbers of employees above and below the average.

This can represent a symmetric distribution.

## AI/ML Example

Understanding symmetry helps in:

- Exploratory Data Analysis
- Statistical modeling
- Distribution analysis
- Selecting suitable statistical methods

## Interpretation

A symmetric distribution does not have a long tail on one side.

The data is approximately balanced around the center.

```python
import numpy as np
import matplotlib.pyplot as plt

data = [10, 20, 30, 40, 50]

plt.hist(data, bins=5)

plt.title("Symmetric Data")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

## Output Interpretation

The values are balanced around the center.

This demonstrates the basic idea of a symmetric distribution.

# 7. Kurtosis

Kurtosis is a statistical measure that describes the tail behavior and the concentration of extreme values in a distribution.

It helps us understand whether a distribution has heavier or lighter tails compared with a normal distribution.

Kurtosis is related to the presence of extreme observations.

## Types of Kurtosis

There are three common types:

1. Mesokurtic
2. Leptokurtic
3. Platykurtic

## 1. Mesokurtic

A normal distribution is commonly used as the reference for mesokurtic behavior.

It has moderate tails.

The excess kurtosis of a normal distribution is approximately 0.

## 2. Leptokurtic

A leptokurtic distribution has heavier tails and more extreme values compared with a normal distribution.

It has positive excess kurtosis.

## 3. Platykurtic

A platykurtic distribution has lighter tails and fewer extreme values compared with a normal distribution.

It has negative excess kurtosis.

## Formula

One common definition of excess kurtosis is:

Excess Kurtosis = Kurtosis - 3

For a normal distribution:

Excess Kurtosis = 0

## Numerical Example

Suppose two datasets have the same mean and standard deviation.

Dataset A has fewer extreme values.

Dataset B has more extreme values.

Dataset B may have higher kurtosis because it has heavier tails.

## Business Example

Kurtosis can be useful in financial analysis.

For example, investment returns with high kurtosis may have a greater tendency to produce extreme returns compared with a normal distribution.

## AI/ML Example

Kurtosis can help during Exploratory Data Analysis to identify features with heavy tails and potential extreme observations.

This can help analysts decide whether further investigation or transformation is required.

## Interpretation

Kurtosis is mainly useful for understanding tail behavior and extreme observations.

Higher kurtosis generally indicates heavier tails and more extreme observations.

Lower kurtosis generally indicates lighter tails and fewer extreme observations.

Kurtosis should not be interpreted simply as a measure of how "peaked" a distribution is; tail behavior is especially important.

```python
import numpy as np
from scipy.stats import kurtosis

data = [10, 12, 13, 14, 15, 16, 17, 18, 20, 50]

kurtosis_value = kurtosis(data)

print("Data:", data)
print("Excess Kurtosis:", kurtosis_value)
```

## Output Interpretation

The kurtosis value provides information about the tail behavior of the dataset.

A positive excess kurtosis suggests heavier tails compared with a normal distribution.

A negative excess kurtosis suggests lighter tails.

A value close to zero is consistent with normal-distribution-level excess kurtosis.

# 8. Empirical Rule (68–95–99.7)

The Empirical Rule describes how data is distributed around the mean in an approximately normal distribution.

According to the Empirical Rule:

- Approximately 68% of the data falls within 1 standard deviation of the mean.
- Approximately 95% of the data falls within 2 standard deviations of the mean.
- Approximately 99.7% of the data falls within 3 standard deviations of the mean.

This is also called the 68–95–99.7 Rule.

## Formula

Approximately 68%:

Mean ± 1 Standard Deviation

Approximately 95%:

Mean ± 2 Standard Deviations

Approximately 99.7%:

Mean ± 3 Standard Deviations

## Numerical Example

Suppose:

Mean = 100

Standard Deviation = 10

### Within 1 Standard Deviation

100 ± 10

Range:

90 to 110

Approximately 68% of observations are expected to fall between 90 and 110.

### Within 2 Standard Deviations

100 ± 20

Range:

80 to 120

Approximately 95% of observations are expected to fall between 80 and 120.

### Within 3 Standard Deviations

100 ± 30

Range:

70 to 130

Approximately 99.7% of observations are expected to fall between 70 and 130.

## Business Example

Suppose a company measures the processing time of customer requests.

Mean processing time = 100 minutes

Standard deviation = 10 minutes

If processing times are approximately normally distributed:

Approximately 68% of requests take between 90 and 110 minutes.

Approximately 95% take between 80 and 120 minutes.

Approximately 99.7% take between 70 and 130 minutes.

## AI/ML Example

The Empirical Rule is useful for:

- Understanding normal distributions
- Identifying unusual observations
- Data analysis
- Statistical modeling
- Understanding z-scores

For example, a value more than 3 standard deviations away from the mean may be considered unusual in an approximately normal dataset.

## Interpretation

The Empirical Rule helps us understand how much data is concentrated around the mean.

It applies specifically to approximately normal distributions and should not be blindly applied to highly skewed distributions.

# Data Distribution Visualization

Visualization helps us understand different types of distributions.

A histogram is useful for observing:

- Center
- Spread
- Shape
- Skewness
- Possible outliers
