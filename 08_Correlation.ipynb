# Correlation & Covariance

# 1. Correlation

## Definition

Correlation is a statistical measure that shows the relationship between two variables.

It tells us:

- The direction of the relationship.
- The strength of the relationship.

The correlation coefficient usually ranges from -1 to +1.

## Correlation Values

- +1 → Perfect positive correlation
- 0 → No linear correlation
- -1 → Perfect negative correlation

## Simple Example

Suppose we have:

Study Hours = [1, 2, 3, 4, 5]

Exam Marks = [50, 55, 65, 75, 85]

As study hours increase, exam marks also increase.

Therefore, they have a positive correlation.

## Numerical Example

Suppose:

X = [1, 2, 3, 4, 5]

Y = [2, 4, 6, 8, 10]

Both variables increase together.

The correlation is:

r = +1

This is a perfect positive correlation.

## Python Implementation

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)
```

## Business Example

A company may study the relationship between:

- Advertising spending
- Sales revenue

If advertising spending increases and sales also increase, there may be a positive correlation.

## AI/ML Example

Correlation can help identify relationships between features and the target variable.

For example:

- House size
- House price

If larger houses generally have higher prices, these variables may have positive correlation.

## Interpretation

A correlation close to +1 means a strong positive relationship.

A correlation close to -1 means a strong negative relationship.

A correlation close to 0 means little or no linear relationship.

## Important Point

Correlation does not prove causation.

If two variables are correlated, it does not automatically mean that one variable causes the other.


# 2. Covariance

## Definition

Covariance measures how two variables change together.

It tells us whether two variables tend to increase or decrease together.

## Formula

Covariance = Σ[(Xi - X̄)(Yi - Ȳ)] / (n - 1)

Where:

Xi = Individual value of X

Yi = Individual value of Y

X̄ = Mean of X

Ȳ = Mean of Y

n = Number of observations

## Covariance Interpretation

- Positive covariance → Variables tend to increase together.
- Negative covariance → One tends to increase when the other decreases.
- Covariance close to 0 → No clear linear relationship.

## Numerical Example

X = [1, 2, 3]

Y = [2, 4, 6]

Mean of X = 2

Mean of Y = 4

Using the covariance formula:

Covariance = 2

Therefore, covariance is positive.

## Python Implementation

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([2, 4, 6])

covariance = np.cov(x, y)[0, 1]

print("Covariance:", covariance)
```

## Business Example

A company can calculate covariance between:

- Advertising expenditure
- Sales revenue

Positive covariance indicates that they tend to increase together.

## AI/ML Example

Covariance is used in Machine Learning and statistics to understand relationships between features.

It is also important in:

- Principal Component Analysis (PCA)
- Multivariate analysis
- Covariance matrices

## Interpretation

The sign of covariance tells us the direction of the relationship.

However, the magnitude of covariance is difficult to compare across variables because it depends on their units.

# 3. Positive Correlation

## Definition

Positive correlation occurs when two variables tend to increase together.

When one variable increases, the other variable also tends to increase.

## Example

Study Hours → Exam Marks

As study hours increase, exam marks generally increase.

## Numerical Example

Study Hours = [1, 2, 3, 4, 5]

Marks = [40, 50, 60, 70, 80]

Both variables increase together.

Correlation ≈ +1

## Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([40, 50, 60, 70, 80])

plt.scatter(x, y)
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.title("Positive Correlation")
plt.show()
```

## Business Example

Advertising expenditure and sales revenue may have a positive correlation.

## AI/ML Example

In a house price dataset:

House Size ↑ → House Price ↑

This may indicate positive correlation.

## Interpretation

The points generally move upward from left to right.

# 4. Negative Correlation

## Definition

Negative correlation occurs when one variable tends to increase while the other variable tends to decrease.

## Example

Exercise Time and Body Weight may show a negative relationship in some datasets.

As exercise increases, body weight may decrease.

## Numerical Example

Exercise Hours = [1, 2, 3, 4, 5]

Weight = [80, 76, 72, 68, 64]

As exercise increases, weight decreases.

Correlation ≈ -1

## Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([80, 76, 72, 68, 64])

plt.scatter(x, y)
plt.xlabel("Exercise Hours")
plt.ylabel("Weight")
plt.title("Negative Correlation")
plt.show()
```

## Business Example

Product price and number of units sold may have a negative correlation.

As price increases, demand may decrease.

## AI/ML Example

In some datasets:

Price ↑ → Demand ↓

This can be a negative relationship between features.

## Interpretation

The points generally move downward from left to right.

# 5. No Correlation

## Definition

No correlation means there is no clear linear relationship between two variables.

The correlation coefficient is close to 0.

## Example

Shoe Size and Exam Marks may have no meaningful relationship.

## Numerical Example

X = [1, 2, 3, 4, 5]

Y = [7, 2, 9, 4, 8]

The values do not follow a clear increasing or decreasing pattern.

Correlation is approximately 0.

## Python Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([7, 2, 9, 4, 8])

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("No Clear Correlation")
plt.show()
```

## Business Example

Customer shoe size and monthly spending may have no meaningful relationship.

## AI/ML Example

If a feature has almost no relationship with the target variable, it may provide limited predictive information.

## Interpretation

The points appear scattered without a clear upward or downward pattern.

## Important Point

A correlation close to zero means there is no strong linear relationship.

It does not always mean that there is absolutely no relationship.

# 6. Pearson Correlation

## Definition

Pearson correlation measures the strength and direction of the linear relationship between two numerical variables.

The Pearson correlation coefficient is represented by r.

## Formula

r = Σ[(Xi - X̄)(Yi - Ȳ)] / √[Σ(Xi - X̄)² × Σ(Yi - Ȳ)²]

## Range

- +1 → Perfect positive correlation
- 0 → No linear correlation
- -1 → Perfect negative correlation

## Numerical Example

X = [1, 2, 3, 4, 5]

Y = [2, 4, 6, 8, 10]

Pearson correlation:

r = +1

This indicates a perfect positive linear relationship.

## Python Implementation

```python
import numpy as np
from scipy.stats import pearsonr

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation, p_value = pearsonr(x, y)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

## Business Example

A company can use Pearson correlation to measure the relationship between advertising spending and sales revenue.

## AI/ML Example

Pearson correlation can be used during feature analysis to identify highly correlated numerical features.

## Interpretation

If Pearson correlation is:

- 0.8 → Strong positive linear relationship
- -0.8 → Strong negative linear relationship
- 0.1 → Very weak linear relationship
- 0 → No linear relationship

## When to Use Pearson Correlation

Use Pearson correlation when:

- Variables are numerical.
- The relationship is approximately linear.
- You want to measure linear correlation.

# 7. Spearman Correlation

## Definition

Spearman correlation measures the strength and direction of a relationship between two variables based on their ranks.

It is useful when the relationship is monotonic but not necessarily linear.

## Formula

ρ = 1 - [6Σd² / n(n² - 1)]

Where:

ρ = Spearman correlation

d = Difference between ranks

n = Number of observations

## Numerical Example

Suppose:

X = [10, 20, 30, 40, 50]

Y = [15, 25, 35, 45, 55]

The ranks of both variables are the same.

Therefore:

Spearman correlation = +1

## Python Implementation

```python
from scipy.stats import spearmanr

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation, p_value = spearmanr(x, y)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

## Business Example

A company can compare:

Customer Satisfaction Rank

and

Customer Loyalty Rank

Spearman correlation can measure whether customers with higher satisfaction rankings also tend to have higher loyalty rankings.

## AI/ML Example

Spearman correlation can be useful when features have an ordered or ranked relationship.

It can also be useful when the data contains outliers or is not normally distributed.

## Interpretation

A Spearman value close to:

+1 → Strong positive monotonic relationship

-1 → Strong negative monotonic relationship

0 → Little or no monotonic relationship

## Pearson vs Spearman

Pearson → Measures linear relationship.

Spearman → Measures monotonic relationship using ranks.

# 8. Correlation Matrix

## Definition

A correlation matrix is a table that shows the correlation coefficients between multiple variables.

Each cell contains the correlation between two variables.

## Example

Suppose we have:

- Study Hours
- Exam Marks
- Attendance

A correlation matrix can show the relationship between all three variables.

## Numerical Example

A correlation matrix may look like:

| | Study Hours | Marks | Attendance |
|---|---:|---:|---:|
| Study Hours | 1.00 | 0.85 | 0.60 |
| Marks | 0.85 | 1.00 | 0.70 |
| Attendance | 0.60 | 0.70 | 1.00 |

## Important Point

The diagonal values are always 1 because each variable has a perfect correlation with itself.

## Python Implementation

```python
import pandas as pd

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Attendance": [60, 65, 70, 80, 90]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

## Business Example

A company can create a correlation matrix for:

- Advertising
- Sales
- Revenue
- Customer Visits

This helps identify relationships between business variables.

## AI/ML Example

A correlation matrix can help identify highly correlated features before training a Machine Learning model.

Highly correlated features may indicate multicollinearity.

## Interpretation

Values close to +1 indicate strong positive relationships.

Values close to -1 indicate strong negative relationships.

Values close to 0 indicate weak or no linear relationships.

# 9. Heatmap

## Definition

A heatmap is a graphical representation of data using different shades or intensities to represent values.

A correlation heatmap displays the correlation matrix visually.

## Why Use a Heatmap?

A heatmap makes it easier to identify:

- Strong positive correlations
- Strong negative correlations
- Weak correlations
- Highly correlated features

## Python Implementation

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Attendance": [60, 65, 70, 80, 90],
    "Assignments": [2, 3, 4, 4, 5]
})

correlation_matrix = data.corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Business Example

A company can use a heatmap to understand relationships between:

- Marketing Spend
- Sales
- Revenue
- Website Visits
- Customer Orders

## AI/ML Example

A heatmap can be used during Exploratory Data Analysis to identify highly correlated features.

For example, if two features have a correlation of 0.95, they may contain very similar information.

## Interpretation

- Value close to +1 → Strong positive correlation
- Value close to -1 → Strong negative correlation
- Value close to 0 → Weak or no linear correlation

The heatmap makes these relationships easier to identify visually.
